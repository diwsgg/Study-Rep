# Unlimited Arguments
>* Lets supposed we have the next function

```python
def add(n1,n2):
    return n1+n2

add(n1=5, n2=3)
```

##### What if i wanted to add more that 2 numbers
**that would be add(n1,n2,n3,.....nm), how we can change that?**

WE wanted to make that function more flexible so we put *args, lets see it


```python
# What is important is the '*' symbol 
# 'args' is by convinence but we can call it whatever name we want

def add(*args):
   for n in args:
      print(n)
```

```python
#Now lets see how we can call it
add(1,2,5,6,9,8,10,23)
```

# Example of sum of any number of arguments

```python
def add(*args):
   return sum(args)

add(1,2,5,6,9,8,10,23)
```

#### **Important**, as we can see, to call the function we do not pass a name for the arguments, and also, is required at least one argument, it can not be empty

##### So we can call the function with names? Or even optional arguments?

# For this we use '**'kwargs arguments

```python
def calculate(**kwargs):
   print(kwargs)
```

```python
#SO how does this will work?

calculate(add=3, multiply=10, c=5)
```

```python
#As we see is a dictionary and the values are
calculate(add=3, multiply=10, c=5)
```

# So what if we wanted to do the operations and not just have that output
## for that we do: 

```python
def calculate(**kwargs):
    sum = 0
    #WE do a for loop, to get the values of each key
    #in this case we do a sum of all the values
    for value in kwargs.values():
      sum += value
    return sum
```

```python
calculate(add=3, multiply=10, c=5)
```

# And as we notice 'add' and 'multiply' arguments are not functions instead only variables names 

# Another example

```python
def calltt(n, **kwargs):
    sum = n
    for value in kwargs.values():
      sum *= value
    return sum
```

```python
calltt(2, add=1, add2=2, mulll=5, fish=10)
```

**So remember**
* *args = without name --> values
* **kwargs = with name --> dictionary


# Example with class

```python
class Car:
    def __init__(self, **kw):
        self.make = kw['make']
        self.model = kw['model']

my_car = Car(make='Nissan', model='GTR-3')
```

```python
print(my_car.make)
print(my_car.model)
```

```python
#If i create a car obj without some value with that defination it will fail
my_car = Car(make='Nissan')
```

```python
#SO to fix this, we add a 'get' instruction
class Car2:
    def __init__(self, **kw):
        self.make = kw.get('make')
        self.model = kw.get('model')

my_car2 = Car2(make='Ford')
```

```python
print(my_car2.make)
print(my_car2.model)
# Even if we do not pass a value, it will execute and will have a None value by default
```

```python
#If we wanted to change that default value we can do
class Car3:
    def __init__(self, **kw):
        self.make = kw.get('make','default Value If Not Pass Any')
        self.model = kw.get('model','Example: new Model 2080')

my_car3 = Car3(make='Toyota', model='Supra')
#So now we will have a default value even if we do not pass any argument
my_car4 = Car3()
```

```python
#with values
print(my_car3.make)
print(my_car3.model)
print('-'*50)
#without any values
print(my_car4.make)
print(my_car4.model)
```
