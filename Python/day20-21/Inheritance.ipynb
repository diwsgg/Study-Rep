# Inheritance

```python
class Animal:
    def __init__(self):
        self.num_eyes = 2

    def breathe(self):
        print("Inhale, exhale.")

#now other class
class Fish:

    def swim(self):
        print("Moving in water")
```

```python
nemo = Fish()
nemo.swim()
```

# Now we want that the class Fish inheritance everything that the class animal can do

```python
class Animal:
    def __init__(self):
        self.num_eyes = 2

    def breathe(self):
        print("Inhale, exhale.")

#We add the class
class Fish(Animal):
    #And in the init we add the super 
    def __init__(self):
        super().__init__()

    def swim(self):
        print("Moving in water")
```

```python
nemo2 = Fish()
nemo2.swim()
nemo2.breathe()
nemo2.num_eyes
```

# We can modify the methods?

```python
class Animal:
    def __init__(self):
        self.num_eyes = 2

    def breathe(self):
        print("Inhale, exhale.")

#We add the class
class Fish(Animal):
    #And in the init we add the super 
    def __init__(self):
        super().__init__()

    #WE can modify breathe?

    def breathess(self):
        super().breathe()
        print("doing it underwater")

    def swim(self):
        print("Moving in water")
```

```python
nemo3 = Fish()
nemo3.swim()
#This is the one from animal
nemo3.breathe()
#This is from the own class
nemo3.breathess()
```

```python
#Given the following:

class Dog:
    def __init__(self):
        self.temperament = "loyal"
 
    def bark(self):
        print("Woof, woof!")
    
#How do you create a class called Labrador (the subclass) that inherits from the Dog class (the superclass)?
```

```python
class Labrador(Dog):
    def __init__(self):
        '''  IMPORTANT 
        
        The call to super() in the initialiser is recommended, but not strictly required
        
        '''
        #SO
        #super().__init__()
        self.temperament = "friendly"

lab = Labrador()
lab.bark()
lab.temperament
```

```python
#What will this code print?

doggo = Dog()
print(f"A dog is {doggo.temperament}")
 
sparky = Labrador()
print(f"Sparky is {sparky.temperament}")
```

```python
#Given this code:

class Dog:
    def __init__(self):
        self.temperament = "loyal"
 
    def bark(self):
        print("Woof, woof!")
 
class Labrador(Dog):
    def __init__(self):
        super().__init__()
        self.is_a_good_boy = True
 
    def bark(self):
        super().bark()
        print("Greetings, good sir. How do you do?")
```

```python
#What will this print?

sparky = Labrador()
sparky.bark()
```
