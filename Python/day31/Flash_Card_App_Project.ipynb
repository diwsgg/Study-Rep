# Step 3 - Flip the Cards!
1. After a delay of 3s (3000ms), the card should flip and display the English translation for the current word.

2. The card image should change to the card_back.png and the text colour should change to white. The title of the card should change to "English" from "French".

```python
def generateWord():
    #We change this variable to make it global and use it for other functions
    global dictvalue, timer_id
    window.after_cancel(timer_id)
    dictvalue = random.choice(dict_words)
    word_randm = dictvalue['French']
    canvas.itemconfig(Languague_text, text= 'French', fill='black')
    canvas.itemconfig(word_text, text=word_randm, fill='black')
    #this will be for the front image
    canvas.itemconfig(canvas_img, image= front_img)
```

```python
def flipCard():
        #make global the variable
        global front, dictvalue, timer_id
        window.after_cancel(timer_id)
        if front:
                canvas.itemconfig(canvas_img, image= back_img)
                canvas.itemconfig(Languague_text, text = 'English', fill='white')
                word_randm = dictvalue['English']
                canvas.itemconfig(word_text, text=word_randm, fill='white')
                front = False
        else:
                canvas.itemconfig(canvas_img, image= front_img)
                canvas.itemconfig(Languague_text, text = 'French', fill='black')
                word_randm = dictvalue['French']
                canvas.itemconfig(word_text, text=word_randm, fill='black')
                front = True
        window.after(ms = 3000, func=flipCard)
```

```python
window = tk.Tk()

window.config(padx=50, pady=50, bg=BACKGROUND_COLOR)
window.title("Flashy")

#canva img
canvas = tk.Canvas(width=800, height=526, bg=BACKGROUND_COLOR ,highlightthickness=0)
front_img = tk.PhotoImage(file=card_front_path)
back_img = tk.PhotoImage(file=card_back_path)

#Initial img
front = True
dictvalue = ''

canvas_img = canvas.create_image(410, 270, image=front_img)
#Display it
canvas.grid(column=0, row=0, columnspan=2)

#labels on canva
Languague_text = canvas.create_text(400, 150, text='', font=("Ariel", 40, "italic"))
word_text = canvas.create_text(400, 236, text='', font=("Ariel", 60, "bold"))

#A first generete word
generateWord()

#WE will use window.after() to check the time pass to flip the card
timer_id = window.after(ms = 3000, func=flipCard)

    
#Button
wrong_image = tk.PhotoImage(file = wrong_path)

ecs_button = tk.Button(text='Search', command=generateWord, image=wrong_image, highlightthickness=0)
ecs_button.grid(column=0, row=2)

right_image = tk.PhotoImage(file = right_path)
check_button = tk.Button(text='Generate Password', command=generateWord, image=right_image, highlightthickness=0)
check_button.grid(column=1, row=2)

window.mainloop()
#we close the process
window.after_cancel(timer_id)
```
