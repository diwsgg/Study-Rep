# What does <Response [200]> MEAN?

* That is our responde code, there are different codes so here is the explanation
* We can identify almost each code with the first digit 
  * Codes
    * 1xx: Hold on, (something is happenging)
    * 2xx: Here you go (is ok)
    * 3xx: You dont have permision to access this
    * 4xx: Does not exists
    * 5xx: server maybe does not exists/work/ or be on maintenance

# All together and make it 60 seconds run every time

```python
import requests
from datetime import datetime
import smtplib

#run every 60 seconds
import time
```

```python
# this will not change 
MY_LAT = 51.507351 # Your latitude
MY_LONG = -0.127758 # Your longitude
email = 'tremaine.gulgowski@ethereal.email'
contr = 'UuvwygRd9SBuEEf2jN' 
ppl= 587
parameters = {
    "lat": MY_LAT,
    "lng": MY_LONG,
    "formatted": 0,
}
```

## Functions

```python
def get_data():

    response = requests.get(url="http://api.open-notify.org/iss-now.json")
    response.raise_for_status()
    data = response.json()

    iss_latitude = float(data["iss_position"]["latitude"])
    iss_longitude = float(data["iss_position"]["longitude"])

    response = requests.get("https://api.sunrise-sunset.org/json", params=parameters)
    response.raise_for_status()
    data = response.json()
    sunrise = int(data["results"]["sunrise"].split("T")[1].split(":")[0])
    sunset = int(data["results"]["sunset"].split("T")[1].split(":")[0])

    time_now = datetime.now()

    return iss_latitude, iss_longitude, sunrise, sunset, time_now
```

```python
def determinate_position(latitude, longitude):
    more_lat = MY_LAT+5
    less_lat = MY_LAT-5

    more_lon = MY_LONG+5
    less_lon = MY_LONG-5

    if latitude >= less_lat and latitude <= more_lat:
        if longitude >= less_lon and longitude <= more_lon:
            return True
    return False
```

```python
def conver_time(current_time):    
    convert_time = str(current_time) # first convert to a stri= (because is an obj)
    #and then split it for only the hours
    convert_time = convert_time.split(' ')[1].split(':')[0]
    return convert_time
```

```python
def send_email():
    with smtplib.SMTP('smtp.ethereal.email', port=ppl) as connection:
        #secure the network
        connection.starttls()

        #we need to login
        connection.login(user= email, password=contr)

        #send the email
        connection.sendmail(from_addr=email, to_addrs='some@gmail.com', msg='Subject: Look Up' \
        '\n\n Its time to look up')
```

```python
#main will run every 60 seconds
def main():
    #first get data
    iss_latitude, iss_longitude, sunrise, sunset, time_now= get_data()
    #then check for position
    if determinate_position(iss_latitude, iss_longitude):
        #convert time
        time_now = conver_time(time_now)
        if time_now>=sunrise and time_now<= sunset:
            send_email()
            print('Email sent')
    print('Email not sent because you are not able to see it')
```

```python
#run the program every 60 seconds
while True:
    #... 60 seconds 
    time.sleep(60)
    main()
```
