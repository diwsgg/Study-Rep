```python
#there are 4 timestamps
print(weather_data['list'][0])
print(weather_data['list'][1])
#the to access the weather data 
print(weather_data['list'][0]['weather'])
#we have a list of just one element so it is always 0
#and the element we wanted is id to compare so
print(weather_data['list'][0]['weather'][0]['id'])
print(weather_data['list'][1]['weather'][0]['id'])
#remember 3 is the maximum index, because goes from 0-3 that is four
print(weather_data['list'][3])
```

```python
#the code for check every id on the data

range_days = len(weather_data['list'])
# it is 4, so the for loop will go from 0 - 4 not inclusive 4
for i in range(range_days):
    #we get the current id
    # i is for the current timestamp, but the 0 is for access the list that weather returns
    current_temp = weather_data['list'][i]['weather'][0]['id']

    if current_temp<700:
        print("Bring a umbrella")
    else:
        print(f'CUrrent temp {current_temp}')
```

```python
import requests
from pathlib import Path
import os
from dotenv import load_dotenv
```

```python
#API_KEYS AND NECESSARY IDs
folder_env = str(Path.cwd().parent)+'/.env'
load_dotenv(folder_env)
API_WEATHER=os.getenv("API_TEMPERATURE_KEY")
API_BOT=os.getenv("API_Telegram_KEY")
ID_BOT=os.getenv("CHAT_ID")
```

```python
data_params = {
    'lat':51.507351,
    'lon':-0.127758,
    'appid':API_WEATHER,
    'cnt':4,
}

response = requests.get(url='https://api.openweathermap.org/data/2.5/forecast',
                        params=data_params)

response.raise_for_status()
weather_data = response.json()
```

```python
for weather in weather_data['list']:
    current_temp = weather['weather'][0]['id']
    if current_temp<700:
        print("Bring a umbrella")
    else:
        print(f'Current temp {current_temp}')
```
