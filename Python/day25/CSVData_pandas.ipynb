# GREAT SQUIRREL CENSUS DATA ANALYSIS

#### Create the dataFrame

```python
#If we use this the problem is that we will include NAN values so
squirrel_data = pd.read_csv('csvdata/2018_Central_Park_Squirrel_Census_-_Squirrel_Data.csv')
```

#### The final csv has to have, fur color, Count

```python
squirrel_data['Primary Fur Color']
```

```python
#We have to count the frequency of the elements
dict_elements = {}

for element in squirrel_data['Primary Fur Color']:
    #this is how we can add a value to a dict
    # dict[key] = dict.get(key,default value (common 0))+1 //+1 will add 1 everytime the key is the same
    dict_elements[element]= dict_elements.get(element, 0)+1

#The problem for this is that we are counting the NAN values too
print(dict_elements)
```

# But we wanted to use pandas library so we use


```python
# We can use the function .value_counts() to count the values of a specific column
counting = squirrel_data['Primary Fur Color'].value_counts()

print(counting)
```

```python
#SOmething interesting is that we can also do this and transform it directly to a list o dictionary
elements = squirrel_data['Primary Fur Color'].value_counts().to_list()
print(elements)
elements = squirrel_data['Primary Fur Color'].value_counts().to_dict()
print(elements)
```

# For last we just have to convert this to a file csv

```python
counting.to_csv("csvdata/squirrel_count.csv")
```
