```python
from flight_data import FlightData
from notification_manager import NotificationManager
from data_manager import DataManager
from flight_search import FlightSearch
```

```python
data_google_sheets = DataManager()
current_Flight = FlightSearch()
```

```python
def comparePrices(data, price : float):
    flights = data.get("best_flights", []) + data.get("other_flights", [])
    if flights: 
        cheapest = min(flights, key=lambda f: f["price"])
        if cheapest['price']<price:
            return cheapest
    return None
```

# SO we have the prject to send notifications to us, but now we will update our project to also have to send emails to the users

```python
def main():
    for flight in data_google_sheets.data_sheety['prices']:

        code = flight['iataCode']
        lowest_price = flight['lowestPrice']

        #compare prices but we need to use first the API for flights
        flight_data = current_Flight.getFlightData(code)
        #comapre
        best_flight = comparePrices(flight_data,lowest_price)
        if best_flight:
            #then we send a notification and an update to the google sheet
            data_google_sheets.PutData(str(flight['id']),best_flight['price'])
            #send
            #first we create an objt 
            flight_info = FlightData(
                departure=current_Flight.departure_id,
                arrival=code,
                price=best_flight['price'],
                outbound_date=best_flight['flights'][0]['departure_airport']['time'].split(" ")[0],
                return_date=best_flight['flights'][-1]['departure_airport']['time'].split(" ")[0],
            )
            #now notification manager it enters
            notification = NotificationManager()
            notification.sendtext(f'Lower Price Alert!, Only ${flight_info.price} to fly from {flight_info.departure} to {flight_info.arrival} on {flight_info.outbound_date} until {flight_info.return_date}')
    print('Process check completed')
```

```python
main()
```

```python
from flight_data import FlightData
from notification_manager import NotificationManager
from data_manager import DataManager
from flight_search import FlightSearch
```

```python
DATA_GOOGLE_SHEETS = DataManager()
CURRENT_FLIGHT = FlightSearch()
```

```python
def newUser():
    name = input("What is your first name: ")
    last_name = input("What is your last name: ")
    email = input("What is your email: ")

    #confirm same email
    temp_email = input("Confirm your email: ") 
    while temp_email != email:
        temp_email = input("Emails does not match\n Put again your email: ")
    #post the data on the data sheet
    DATA_GOOGLE_SHEETS.PostUserData(name, last_name, email)
    #and update the data for the data users
    DATA_GOOGLE_SHEETS.getUsersData()
    #we do not have to return anything and we are updating the sheet
```

```python
def comparePrices(current_flight, current_lowerprice : float):
    if current_flight: 
        cheapest = min(current_flight, key=lambda f: f["price"])
        if cheapest['price']<current_lowerprice:
            return cheapest
    return None
```

```python
def convert(data):
    flights = data.get("best_flights", []) + data.get("other_flights", [])
    return flights
```

```python
def flight():
    for flight in DATA_GOOGLE_SHEETS.data_sheety['prices']:
        
        code = flight['iataCode']
        lowest_price = flight['lowestPrice']
        stop = 1
        
        data_flight = CURRENT_FLIGHT.getFlightData(code=code, stops=stop)
        #transform the data
        flights = convert(data=data_flight)

        #check and if flight is empty then get again another data without stops
        while stop==1 and not flights:
            stop=None
            data_flight = CURRENT_FLIGHT.getFlightData(code = code, stops=stop)
            #transform the data
            flights = convert(data=data_flight)

        #and check the cheapest
        cheapest_flight = comparePrices(flights, lowest_price)

        if cheapest_flight is not None:
            print("We found a deal!")

            id = str(flight['id'])
            new_price = cheapest_flight['price']

            DATA_GOOGLE_SHEETS.PutData(id, new_price)

            #first we create an objt 
            flight_info = FlightData(
                departure=CURRENT_FLIGHT.departure_id,
                arrival=code,
                price=new_price,
                outbound_date=data_flight['search_parameters']['outbound_date'],
                return_date=data_flight['search_parameters']['return_date'],
                stops=len(cheapest_flight["flights"]) - 1
            )
            #now notification manager it enters
            notification = NotificationManager()
            #create the letter
            letter = f'Lower Price Alert!, Only ${flight_info.price} to fly from \
                       {flight_info.departure} to {flight_info.arrival} on {flight_info.outbound_date} until {flight_info.return_date}'
            
            #send to telegram
            notification.sendtext(letter)

            #send to every user
            for user in DATA_GOOGLE_SHEETS.data_users['users']:
                notification.sendEmail(user['whatIsYouEmail?'], letter)

        else:
            print(f"There is not deal for {CURRENT_FLIGHT.departure_id} to {code}")

    print("Done, Check for all flights of today")
```

```python
def menu():
    option = int(input('1. Are you a new user, Registred\n'\
                   '2. Search for sales\n'\
                   '3.Out\n'))
    return option
```

```python
def main():
    option = 0
    while option!=3:
       option = menu()
       print(option)
       match option:
            case 1:
                newUser()
            case 2:
                flight()
            case 3:
                print("Have a nice day")
            case _:
                print("No correct option, try again\n")   
```

```python
main()
```
