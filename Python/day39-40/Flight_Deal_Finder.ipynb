```python
# well for that we have to use the data manager class so first we will do a test3
import requests
from pathlib import Path
import os
from dotenv import load_dotenv
from datetime import datetime, timedelta
```

```python

#the path for the ENV
#FOR PY FILE env_path = str(Path.cwd())+'/Python/.env'
#FOR JUPYTER
env_path = str(Path.cwd().parent)+'/.env'
load_dotenv(env_path)
API_KEY = os.getenv('Flight_APIKEY')
ENDPOINT = "https://app.100daysofpython.dev/v1/flights/search"
```

```python
tomorrow         = datetime.now() + timedelta(days=1)
six_month_later  = datetime.now() + timedelta(days=6 * 30)
```

```python
#as we can see with timedelta function we do not have to worry about sum a day o 180 and check for year, month
print(tomorrow)
print(six_month_later)
```

```python
params = {
    "api_key":       API_KEY,
    "engine":        "google_flights",                      #this is always like that
    "departure_id":  "LHR",                                 #IATA CODE
    "arrival_id":    "CDG",                                 #IATA CODE
    "outbound_date": tomorrow.strftime("%Y-%m-%d"),         #THE FORMAT FOR DATE IS YYYY-mm-dd
    "return_date":   six_month_later.strftime("%Y-%m-%d"),  #same format
    "type":          "1",                                   #1 (round trip) or 2 (one way).
    #----------------------------------------OPTIONAL----------------------------------------
    "adults":        "1",                                   #DEFAULT 1
    "currency":      "GBP",                                 #ONLY SUPPORTED GBP
    "stops":         "1",                                   #1 = NONSTOP ONLY, 
}
# we can not have a header params, since the API says that is necessary on the params
header_params ={
    "x-app-key":       API_KEY,
}
```

```python
response = requests.get(ENDPOINT, params=params)
data = response.json()

if "error" in data:
    print("API error:", data["error"])
else:
    flights = data.get("best_flights", []) + data.get("other_flights", [])
    cheapest = min(flights, key=lambda f: f["price"])
    print(f"Cheapest LHR-CDG: GBP {cheapest['price']}")
```

```python
cheapest = min(flights, key=lambda f: f["price"])

#we have to remember that min takes
#minimun = min(iterable, key)
#the key is when we have not only integers on the iterable vairable, like in this case a dictionary
#we will have an error if we only left flights, so
#what about the key lambda

lambda f: f["price"] #extracts the price from each dictionary.
#lets see it
```

### How does lambda work?
> A lambda is an anonymous function: **a function defined without using def.**

Its syntax is:
>lambda **parameters**: **expression**

In this case we have:
> lambda **f**: **f["price"]**

is equivalent to:
>def get_price(f):    return f["price"]

Therefore, these two expressions behave the same way:
> cheapest = min(flights, key=lambda f: f["price"])

> cheapest = min(flights, key=get_price)

* Notice something important:
>we write key=get_price, not key=get_price().

Because min() needs the function itself so it can call it for each element.

# Now use Sheety API to read Airpot and Price data 

```python
# well for that we have to use the data manager class so first we will do a test3
import requests
from pathlib import Path
import os
from dotenv import load_dotenv

#the path for the ENV
#FOR PY FILE env_path = str(Path.cwd())+'/Python/.env'
#FOR JUPYTER
env_path = str(Path.cwd().parent)+'/.env'
load_dotenv(env_path)
API_KEY_SHEETY = os.getenv('AUTHORIZATION_FLIGHTSheety')
ENDPOINT_SHEETY = "https://api.sheety.co/56df9d1d9fe8908c5a3bcfe5d729cf6b/iataCodes/prices"

header_sheety={
    'authorization':API_KEY_SHEETY,
}
```

```python
response_sheety = requests.get(url=ENDPOINT_SHEETY, headers=header_sheety)

response_sheety.raise_for_status()

data_sheety = response_sheety.json()
```

```python
#with this we have the list of all the data on our sheet
print(data_sheety['prices'])
```

### We wanted the IATACODE, and the price, then we can move on on the next city and get the same 

```python
for row in data_sheety['prices']:
    print(row)
```

```python
#for later in the class
# we can make those attributes, to have them in a list of objects
#where we can intialize them with the sheet data

codes_prices = []
for row in data_sheety['prices']:
    current_code = row['iataCode']
    current_lowestPrice = row['lowestPrice']
    #those data need to be one object with that the objt will work
    codes_prices.append((current_code,current_lowestPrice))

print(codes_prices)

print(codes_prices[0])
#this could be represented as an attribute for
#codes_prices[0].code
#or
#codes_prices[0].price
#like this
print(codes_prices[0][0])
    
```

```python
#now for do a request with the api of flight we can do this
for flight in codes_prices:
    params = {
        "arrival_id":    flight[0], #this is the data we change
    }
    print(params)
```
