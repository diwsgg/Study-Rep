```python
def getTodayDate():
    today_date = (str(dt.datetime.now())[:10]).split('-')
    return today_date
```

```python
today_date = getTodayDate()
print(today_date)
```

```python
def transformDate(date):
    #Get the data divided and tramsformed it
    year = int(date[0])
    month = date[1]
    day = int(date[2])

    return year, month, day
```

```python
year, month, day = transformDate(today_date)
print(year,month,day)
```

```python
def restDays(days : int):
    yesterday = days-1
    yesterdayDB = days-2
    #transform it to strings
    yesterday = str(yesterday)
    yesterdayDB = str(yesterdayDB)
    return yesterday, yesterdayDB
```

```python
def newKeys(year, month, yesterday, yesterdayDB):

    date_yesterday = str(year)+'-'+str(month)+'-'+str(yesterday)
    date_yesterdayDB = str(year)+'-'+str(month)+'-'+str(yesterdayDB)

    return date_yesterday, date_yesterdayDB
```

```python
def check_days(day: int, month: str, year : int):
    list_months = {'04','06','09','11'}
    temp_months = int(month)
    if day==1: 
        if temp_months>1:
            if month in list_months:
                day=31
            else:
                day=30
            #dinamic typing
            temp_months-=1
        else:
            #dinamic typing
            month='12'
            day=31
            #this the year will have 
            year-=1
    #transform the months again
    if month!='12' and temp_months<10:
        month='0'+str(temp_months)

    #Get yesterday and day before yesterday
    yest, yyest = restDays(day)

    #and generate 2 new keys
    Key_One, Key_Two = newKeys(year, month, yest, yyest)
    
    return Key_One, Key_Two
```

```python
KeyOne, KeyTwo = check_days(day, month, year)
```

```python
print(KeyOne, KeyTwo)
```

```python
# STOCK price yesterday
print(data['Time Series (Daily)'][KeyOne])

print('-'*120)

# STOCK price day before yesterday
print(data['Time Series (Daily)'][KeyTwo])
```

## **Whats next?**
From step one we need to verify
* When STOCK price increase/decreases by 5% between yesterday and the day before yesterday then print("Get News").
  
We know the data from the two days, but how we can defined that it was an increase or a decrease? 
* we just need the 'Closing price'


**[FORMULA]**

> Percentage Change = ((New Price - Old Price) / old Price) * 100


```python
def PercentageChange(data, keyOne, KeyTwo):
    closing_yesterday = float(data['Time Series (Daily)'][keyOne]['4. close'])
    closing_DByesterday = float(data['Time Series (Daily)'][KeyTwo]['4. close'])

    percentage_change = round(((closing_yesterday - closing_DByesterday) / closing_DByesterday) * 100,2)

    return percentage_change
```

```python
percentage_change = PercentageChange(data,KeyOne, KeyTwo)
print(percentage_change)
```

```python
if percentage_change>=5 or percentage_change<=-5:
    print('Get News')
else:
    print(f'change is not significative:{percentage_change}%')    
```

```python
#Data for stock market
def getData():
    response_data = requests.get(url=api_link, params=parameters)
    response_data.raise_for_status()
    #we know that the data we wanted is for the dictionary Time Series (Daily), so 
    data = response_data.json()['Time Series (Daily)']
    return data
```

```python
def transform_data_stock(data):
    data_list = [value for (_,value) in data.items()]
    yesterday = data_list[1]
    DB_yesterday = data_list[2]
    cost_yesterday = float(yesterday['4. close'])
    cost_DB_yesterday = float(DB_yesterday['4. close'])
    return cost_yesterday, cost_DB_yesterday
```

```python
def PercentageChange(cost_yesterday, cost_DB_yesterday):
    #some important change we really do not care right now for the negative or positive 
    #(that is from our professor)
    percentage_change = round((abs(cost_yesterday - cost_DB_yesterday) / cost_DB_yesterday) * 100,2)
    return percentage_change
```

```python
#send the message for telegram
def telegram_bot_sendtext(bot_message):
    bot_token = 'API_BOT'
    bot_chatID = 'ID_BOT'
    send_text = 'https://api.telegram.org/bot' + bot_token + '/sendMessage?chat_id=' + bot_chatID + '&parse_mode=Markdown&text=' + bot_message

    response = requests.get(send_text)

    return response.raise_for_status()
```
