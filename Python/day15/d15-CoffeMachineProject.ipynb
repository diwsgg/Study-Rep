```python
# For a new execution need to restart the kernel, if you dont do it, then the variables from data will keep the same after the 
# operations done
from data import MENU, resources, options_to_pay
```

```python
def checkIngredients(optionDrink):
    for option, amountIngredients in MENU[optionDrink]["ingredients"].items():
        if resources[option] < amountIngredients:
            print("Sorry not enough ingredients...")
            return False
    return True
```

```python
def getAmountMoney():
    subtotal = 0
    total = 0
    for option, amount in options_to_pay.items():
        print(f"How many {option}: ")
        subtotal = int(input("How many (1,2,3,....)"))
        subtotal = amount* subtotal
        total += subtotal
    return total 
```

```python
def checkPayment(optionDrink, money):
    total = getAmountMoney()
    if total>MENU[optionDrink]["cost"]:
        total -= MENU[optionDrink]["cost"]
        print(f"Here is your change: ${round(total)}")
        money.append(MENU[optionDrink]["cost"]) 
        return True
    elif total<MENU[optionDrink]["cost"]:
        print("Sorry not enough money, ")
        return False
    else: 
        print(f"Exact amount")
        money.append(MENU[optionDrink]["cost"]) 
        return True
```

```python
#Now do the coffe
def MakeCoffe(optionDrink, money):
    #First check about the ingredients
    if checkIngredients(optionDrink):
        #Then if its true we know make the payment
        if checkPayment(optionDrink, money):
            #If all pass then we do the operation to rest the ingredients
            for option, amount in MENU[optionDrink]["ingredients"].items():
                resources[option] -= amount
            print(f"Enjoy your {optionDrink}") 
```

```python
def report(money):
    print(f'WATER: {resources["water"]}ml')
    print(f'Milk: {resources["milk"]}ml')
    print(f'Coffe: {resources["coffe"]}g')
    print(f"Money: ${sum(money)}")
```

```python
def Displaymenu():
    print("Costs of Drinks:")
    #Just cost
    print(f"Espresso, cost: ${MENU["espresso"]["cost"]}")
    #Just cost
    print(f"Latte, cost: ${MENU["latte"]["cost"]}")
    #Just cost
    print(f"Capucchino, cost: ${MENU["capucchino"]["cost"]}")
```

```python
#Dictionary for operations
operations = {
    "report" : report,
    "espresso" : MakeCoffe,
    "latte" : MakeCoffe,
    "capucchino" : MakeCoffe,
    "off" : True
}
```

```python
def main():
    print("Welcome to the Coffe Machine")
    
    #Variables
    options_to_drink = ''
    end = False
    #This will be used for tell to the user in the report how much gains it has
    #AS list are mutable we can make changes without passing the variable
    #At then end with report function we can only du sum(money...)
    money_in_machine = []

    while not end: #and options_to_drink != 'off' 
        print("What would you like? (espresso/latte/capucchino): ")
        print("-"*50)
        Displaymenu()
        print("-"*50)
        options_to_drink = input('espresso / latte / capucchino').lower()
        #Check the option
        if options_to_drink == 'espresso' or options_to_drink == 'latte' or options_to_drink == 'capucchino':
            operations[options_to_drink](options_to_drink, money_in_machine)
            print("-"*50)
        elif options_to_drink =='report':
            print("REPORT: \nRemaining Ingredients and Money on machine\n")
            operations[options_to_drink](money_in_machine)
            print("-"*50)
        else:
            end = operations[options_to_drink]
    print("Have a nice day")
```

```python
main()
```
