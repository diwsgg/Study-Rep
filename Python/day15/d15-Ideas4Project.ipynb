# Now how we can compare if the result is negative then we do not want to do that operation and restore the result

```python
# Now we are resting the amount in the dictionary resources
resources = {
    "water": 300,
    "milk": 200,
    "coffe": 100,
}

option2 = "latte"
```

```python
#WE can check first if there will be enough ingredients for make the rest how we can now if there will be enough
# IF 
#AMOUNT(RESOURCES[VALUE]< AMOUNT_NEEDED) THEN IS NOT POSSIBLE TO MAKE THE REST, IF ALL OF THEM ARE TRUE THEN WE CAN DO IT
def checkingredients(menuOption):
    for option, amount in MENU[menuOption]["ingredients"].items():
        if resources[option] < amount:
            print(f"Sorry, there is not enough {option}.")
            return False
    return True
```

```python
if checkingredients(option2):
    for option, amount in MENU[option2]["ingredients"].items():
        resources[option] -= amount
    print(resources)
```

# Now, we can check if we have enough ingredients, we can make the substract to the machine, now we need the "money" of the user

```python
#There are for options

# penny = 0.01
# dime = 0.10
# nickel = 0.05
# quarter = 0.25

options_to_pay={
    'penny' : 0.01,
    'dime' : 0.10,
    'nickel' : 0.05,
    'quarter' : 0.25,
}


#WE can just do a multiplication for that and calculate if the user put more money and gave the rest of that, 
# if is not enough then we cant do the sale, and the money is "refund"
#So we have to  check first if we have enough ingredients and then check for the payment 

#So after this
#checkingredients(option2)
total = 0
subtotal = 0
#WE are checking for the money
for option, amount in options_to_pay.items():
    print(f"How many {option}?: ")
    selected = int(input("How many coins (as 1,2,3,4)"))
    print(selected)
    subtotal = amount*selected
    total += subtotal

#After the total, we are checking
if total>MENU[option2]["cost"] :
    print("Perfect here is your change: ")
    total -= MENU[option2]["cost"]
    print(total)
    #ALSO WE ADD THE RESULT TO A NEW VARIABLE CALL MONEY
elif total<MENU[option2]["cost"]:
    print("Not enough")
else:
    print("perfect")
    
print(total) 
```
