# BMI Calculator with Interpretations
> Add some if/elif/else statements to the BMI calculator so that it interprets the BMI values calculated.

* If the bmi is under 18.5 (not including), print out "underweight"

* If the bmi is between 18.5 (including) and 25 (not including), print out "normal weight"

* If the bmi is 25 (including) or over, print out "overweight"

```python
#Variables
weight = 85
height = 1.85

bmi = weight / (height ** 2)
```

```python
if bmi < 18.5:
    print("underweight")
elif bmi>=18.5 and bmi<25: #USE 'AND' because we need to satisfied both conditions, not only one
    print("normal weight") 
else:
    print("overweight")
```

# Python Pizza

Our job is to build an automatic pizza order program
BASed on a user's order, work out their final bill.

> Use the input() function to get a user's preferences and then add up the total for their order and tell them how much they have to pay.
* Small pizza (S): $15
* Medium pizza (M): $20
* Large pizza (L): $25
* Add pepperoni for small pizza (Y or N): +2
* Add pepperoni for medium or large pizza (Y or N): +3
* Extra chesse (Y or N): +1

```python
print("Welcome to Python Pizza Delivery")
#First determinate the size of the pizza
type_pizza = input("What size pizza do you want? (S, M, L)")
subtotal=0
if type_pizza == 'S':
    subtotal += 15
elif type_pizza == 'M':
    subtotal += 20
elif type_pizza == 'L':
    subtotal += 25
else:
    print("Not a correct option")

#Then asked for extra pepperoni
answ1 = input("Do you want pepperoni on your pizza? (Y or N)") #----Only for small
if answ1 == 'Y':
    if type_pizza == 'S':
        subtotal +=2
    else:
        subtotal +=3
else:
    print("Is a not")

#Then asked for extra chees
answ2 = input("Do you want extra cheese on your pizza? (Y or N)") # FOR ALL
if answ2 == 'Y':
    subtotal +=1
else:
    print("Is a not")
#The final_total, could work if you wanted to add a different operation or make it more cleaner
final_total = subtotal
```

```python
print(f"Your total is {final_total} dollars, Thanks for buying.")
```
