# LISTS

**Lists can be created as:**

>fruits = [item1, item2]

* The lists can be fill with strings, integers, booleans
* Examples

```python
#Strings
fruits = ["Cherry", "Apple", "Pear"]

#No need a loop to print every item on the list
print(fruits) # All the data
print(fruits[0]) # The first element
print(fruits[-1]) # The last element (works backwards)
```

# More options with lists

Yes we have more as
> * .extend(iterable)
* Can be a new list
> * .insert(i, x)
* Insert an item in the position we gave
> * .remove(x)
* Remove the first item on the list whose value is equal to X 
> * .pop([i])
* Remove the item at the given position and return it
* If no "i" is specified, then it will remove and returns the last item in the list 

**This are some of them, and for more information is on dcs.python.org/3/tutorial/datastrcutures.html** 
# Examples

#### Extended

```python
list = ["item1", "item2", "item3", "item4"]

#EXTEND

list.extend(["item5","item6"])
print(list)
```

#### Insert

```python
list = ["item1", "item2", "item3", "item4"]

#Insert

list.insert(1,"item002")
print(list)
```

#### Remove

```python
list = ["item1", "item2", "item3", "item4"]

#Remove

list.remove("item3")
print(list)
```

#### POP

```python
list = ["item1", "item2", "item3", "item4"]

#POP

list.pop(1) #with index
print(list)

list.pop()
print(list) #without index (the last item)
```
