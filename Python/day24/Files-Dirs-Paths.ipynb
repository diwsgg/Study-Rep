# Files, DIrectories and Paths

```python
#SO we have one file on our directory, how we can open it

#We have to declare a variable for open the file
file  = open("file.txt")
#Then created another variable to get the content of that file 
content = file.read()

print(content)
#This will not print the content
#print(file.read())

#We need to close the file

file.close()
```

```python
#Another way to do read a file, without closing it is:
with open("file.txt") as file:
    content = file.read()

print(content)
```

```python
#If we want to write something new in the file we have to use
#BUT THIS WILL DELETE THE CONTENT AND WILL REPLACE IT WITH THIS
with open("file.txt","w") as file:
    file.write("Something new")

#This W instruction can create a file if does not exist     
```

```python
#If we wanted to add something without deleting the content
#a is for append
with open("file.txt","a") as file:
    file.write("\nSomething new")
    #content = file.read()

#For the operation we can only do one a time
```

```python
#We define the paths
from pathlib import Path
# folder_path = Path(__file__).parent
#JUPYTER
folder_path = Path.cwd()

names_path = folder_path / "Input"/ "Names" / "invited_names.txt"

starting_path = folder_path / "Input"/ "Letters" / "starting_letter.txt"
```

# Start with reading the names

```python
with open(names_path,"r") as names:
    #We can use readlines
    #This will return a list, but the problem is that this will have an \n at the end 
    file_names = names.readlines()
print(file_names)
```

```python
#SO an alternative
with open(names_path,"r") as names:
    #We can use read and splitlines
    file_names = names.read().splitlines()
print(file_names)
```

# WE can also use a comprehension list with .readlines()
* Using 
  * .rstrip()
    * remove what we put inside the function
  * .strip()
    * remove all leading/trailing whitespace including newlines

```python
with open(names_path,"r") as names:
    file_names = [line.rstrip('\n') for line in names.readlines()]
print(file_names)
```

```python
with open(names_path,"r") as names:
    file_names = [line.strip() for line in names.readlines()]
print(file_names)
```

# Now we need the file for the start letter

```python
#WE need the file of the startletter
with open(starting_path,"r") as starting:
    file_letter = starting.read()

print (file_letter)
```

# How we can replace the [name] with our names 

```python
#test to replace
newletter = file_letter

newletter = newletter.replace("[name]", file_names[0])

print(newletter)
```
