```python
def strStr(haystack: str, needle: str) -> int:
    a = len(haystack) 
    # We do not need to check anything else
    # but the lenght of the two strings
    # is needle is larger then that means haystack can not have the word 
    # so the result is -1, and what if are the same 
    # but has no coincidences well in that case also we return -1
    #IMPORTANT
    # A '>=' b since 1>1 false, so we need to make it 1>=1 true to check that
    # why not return 0 because we are not sure if the two strings are the same, so we need to check that
    # for other things such as needle>haystack is not possible to get the word so thats is a -1 return  
    if a >= len(needle):
        for i in range(a):
            if haystack[i:i+len(needle)] == needle:
                return i
        return -1
    return -1
```

# Solution

```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        #WE can make a comparation if we do not wanted to assign a varaible
        a = len(haystack) 
        if a >= len(needle):
            #The assignation is for not make a double calculation for the loop
            for i in range(a):
                #Same here for not make a variable call substring
                #but we can do it if we want it more especific
                #and each substring we are only changing one index at time, we do not need two pointers
                #no more if statements since if all substrings are not needle then there is no index tu return 
                if haystack[i:i+len(needle)] == needle:
                    return i
            return -1
        return -1

sol = Solution()
```

```python
print(sol.strStr('butsad','sad'))
print(sol.strStr('leetcode','leeto'))
print(sol.strStr('sadbutsad','sad'))
print(sol.strStr('aaa','aaaa'))
print(sol.strStr('a','a'))
```

# An improvement for the loop range

> **for i in range(a - len(needle) + 1):**
 
How this works

* a = lenght of **handle** = 6 = 'ABCDEF'
* lenght of **needly** = 3 = 'CDE'

> Now

* 6 - 3 = 3
But why 3 well

```python
'''
Indexes:
0   1   2   3   4   5
-------------------------
a   b   c   d   e   f

Index 0:
       0   1   2
       ---------
       a   b   c

Index 1:

           1   2   3
           ---------
           b   c   d

Index 2:

               2   3   4
               ---------
               c   d   e

Index 3:

                   3   4   5
                   ---------
                   d   e   f

But Index 4:

                4   5   ?
                ---------
                e   f
       
WE can't create from index 4 to beyond any substring of lenght of 'needle', that why we only can go from 0 to 3
now .... 
'''
```

**Thats why**
> 3 = len(haystack) - len(needle)
>  * = 6 - 3

* But why the +1

SO if we remember the range() function in python does not include the final number [1,2,3,... n] the range will be from 0-n-1 == that means if we wanted to include the last substring valid, in this case 3, we have to put a +1

> range(6-3+1) --> range( from 0 to 3 ) --> range(4) and is correct

```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        #WE can make a comparation if we do not wanted to assign a varaible
        a = len(haystack) 
        b = len(needle)
        if a >= b:
            for i in range(a-b+1):
                if haystack[i:i+len(needle)] == needle:
                    return i
            return -1
        return -1

sol = Solution()
```
