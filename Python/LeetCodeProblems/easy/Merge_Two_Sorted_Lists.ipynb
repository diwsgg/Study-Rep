```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def mergeTwoLists(self, list1: ListNode | None, list2: ListNode | None) -> ListNode | None:
        listTemp = ListNode()
        current = listTemp
        while list1 and list2:
            if list1.val < list2.val:
                current.next = list1
                list1 = list1.next
            else:
                current.next = list2
                list2 = list2.next
            current = current.next
        if list1:
            current.next = list1
        else:
            current.next = list2
        return listTemp.next
```

```python
#Define the nodes
node1 = ListNode(1)
node2 = ListNode(2)
node3 = ListNode(4)

nodeN1 = ListNode(1)
nodeN3 = ListNode(3)
nodeN4 = ListNode(4)
#Assign the next value
node1.next = node2
node2.next = node3

nodeN1.next = nodeN3
nodeN3.next = nodeN4

list1 = node1
list2 = nodeN1

solFinal = Solution()

print(solFinal.mergeTwoLists(list1, list2))
```

# Important things

>'*while list1 and list2:*' 
* **is the same as** 
>'*while list1 is not None and list2 is not None:*'
  
*  **The two pointers are important** even if we just think one is perfect, The reason:
   *  **listTemp** = remembers the beginning
   *  **current** = moves through the list while we're building it



```python
'''
listTemp
   ↓
[dummy] → [1] → [1] → [2] → [3]
                             ↑
                           current
'''
```

At the end of our loop

```python
'''                           
listTemp
   ↓
dummy → 1 → 1 → 2 → 3 → 4 → 4
                              ↑
                            current
'''
```

if we return listTemp.next the objt will have

```python
'''
1 → 1 → 2 → 3 → 4 → 4
'''
```

* if we only work with current
> the current value at the end will be none and next does not even exist also

> **That's the key.**

* You need two concepts:

1. Where does my merged list START?
2. Where am I CURRENTLY adding nodes?

> The two pointers represent those two jobs:

* listTemp = ListNode()--->   **remembers the start**
* current = listTemp --->     **moves while building**
