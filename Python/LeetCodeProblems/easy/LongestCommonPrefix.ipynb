# Comparing the characters

```python
def longestCommonPrefix(strs: list[str]) -> str:
    prefix = ""
    #SO we know that the maximum Index will be the shortest string  so
    for i in range(len(min(strs, key=len))):
        #This loop will be from 1 to n-1 
        for j in range(1, len(strs)):
            #compare with first string's characters
            if strs[j][i] != strs[0][i]:
                #if a character of the other strings are different then we return the prefix that we have
                return prefix
        #If we put this outside j loop the it will do it i times 
        prefix += strs[j][i]

    return prefix


print(longestCommonPrefix(strs = ["flower","flow","flight"]))
print(longestCommonPrefix(strs = ["dog","racecar","car"]))
```

# Solution

```python
class Solution:
    def longestCommonPrefix(self, strs: list[str]) -> str:
        prefix = ""
        for i in range(len(min(strs, key=len))):
            for j in range(1, len(strs)):
                if strs[j][i] != strs[0][i]:
                    return prefix
            prefix += strs[0][i]
        return prefix

sol = Solution()
```

```python
print(sol.longestCommonPrefix(strs = ["flower","flow","flight"]))
print(sol.longestCommonPrefix(strs = ["dog","racecar","car"]))
```

# for best time

```python
class Solution:
    def longestCommonPrefix(self, strs: list[str]) -> str:
        prefix = ""
        shortest_String = min(strs, key=len)
        for i in range(len(shortest_String)):
            for word in strs:
                if word[i] != shortest_String[i]:
                    return prefix
            prefix += shortest_String[i]
        return prefix

sol = Solution()
```

```python
print(sol.longestCommonPrefix(strs = ["flower","flow","flight"]))
print(sol.longestCommonPrefix(strs = ["dog","racecar","car"]))
```
