```python
def removeElement(nums: list[int], val: int) -> int:
        if len(nums)<1 and val not in nums:
                return len(nums)
        else:
            #we do not wanted the val on the list
            k=0
            for index, num in enumerate(nums):
                print(index, num)
                if num != val:
                    k+=1
                else:
                      nums[index] = -1 
            # it is say that does not matter the order
            # but first is going to be the k elements that are are equal to -1
            nums.sort(reverse=True)
            print(nums)
            return k
```

```python
class Solution:
    def removeElement(self, nums: list[int], val: int) -> int:
        #counter of unique elements
        w = 0
        #we iterate with an index
        for r in range(len(nums)):
            #if the number is diferent from the value
            if nums[r] != val:
                nums[w] = nums[r]
                w = w + 1
        return w
```
