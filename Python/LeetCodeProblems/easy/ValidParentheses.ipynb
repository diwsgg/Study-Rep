```python
def isValid(s: str) -> bool:

    hashmap = {
        '(': ')',
        '{': '}',
        '[': ']',
    }

    if (len(s)==1):
        return False
    
    stack = []

    for p in s:
        if p in hashmap:
            stack.append(p)
        elif stack !=[]:
            if hashmap[stack[-1]] == p:
                stack.pop()
            else:
                return False
        else:
            return False

    if stack==[]:
        return True
    else:
        return False
```

```python
#from collections import deque


def isValid(s: str) -> bool:

    # means double-ended queue.
    # It is designed to efficiently add/remove elements from either end.
    #stack = deque()
    #We actually do not need deque(), since we are only needing
    # push and pop from right (top list)
    stack = []

    #This reverse the hasmap logic that we did
    map = {
        ")" : "(",
        "}" : "{",
        "]" : "[",
    }
    #THIS SAYS, if character is ) then i have to close it with )
    # And we are comparing that directly agains the top of the stack

    for c in s:
        #if character is on map
        if c in map:
            # We are checking that 
            # first stack is not empty
                # An empty collection is considered False.
                #stack=[] --> false
                # not stack[] --> true

                #not stack[1,2,3]--> false
            #Second stat.pop()!= map[c] what does this means

            #Consider:
                # s = "(]"
                # After processing '(':
                # stack = ["("]
                # Now:
                # c = "]"
                # Since ] is in map: 
                # map[c] --> [
                # And:
                # stack.pop()
                # returns:
                # "(" and simultaneously removing it.
                #now for last
                # not stack --> True
            # Thats why we first check that stack is not empty  
            # for python 'or ...' is not evaluated if the first part is true  
            if not stack or stack.pop() != map[c]:
                return False
        #If is not an end bracket we append that to the list
        else:
            stack.append(c)

    #At last
    # if stack is not empty we will return False
    # if is empty return True, that is what
    #'not stack' gave us at same time for not use 2 if statements
    return not stack
```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        map = {
            ")" : "(",
            "}" : "{",
            "]" : "[",
        }

        for c in s:
            if c in map:
                if not stack or stack.pop() != map[c]:
                    return False
            else:
                stack.append(c)
        return not stack
```
