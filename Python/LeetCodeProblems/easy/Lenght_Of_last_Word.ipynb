# Length of Last Word

Given a string s consisting of words and spaces, return the length of the last word in the string.

A word is a maximal substring consisting of non-space characters only. 

> Example 1:

* Input: s = "Hello World"
* Output: 5
* Explanation: The last word is "World" with length 5.
> Example 2:

* Input: s = "   fly me   to   the moon  "
* Output: 4
* Explanation: The last word is "moon" with length 4.
> Example 3:

* Input: s = "luffy is still joyboy"
* Output: 6
* Explanation: The last word is "joyboy" with length 6.
 

> Constraints:

* 1 <= s.length <= 104
* s consists of only English letters and spaces ' '.
* There will be at least one word in s.

```python
class Solution:
    def lengthOfLastWord(self, s: str) -> int:
        s = s[::-1]
        counter = 0
        #we will make a loop 
        for c in s:
            #compare only the last word
            if c !=' ':
                counter+=1
            else:
                break

        return counter
```

# So the problem comes when "   fly me   to   the moon  " the last thing on the string is an space, how we can resolve that

```python
# we do not need to start from the beginning we actually only need to start from the end

def lengthOfLastWord(s: str) -> int:
    #counter
    counter = 0

    for i in range(len(s)-1,0,-1):
       #compare
        if counter!=0 and s[i]==' ':
           return counter
        elif s[i]!=' ':
            counter+=1
```

# That works but another solution could be this one, (this is called pythonic way, since we use python functions to do it faster)

```python
def lengthOfLastWord(s: str) -> int:
    #so we can have a list of words with the function split
    words = s.split()
    print(words)
    #if the list is empty
    if not words:
        return 0
    #if not we will return the lenght of the last item
    #we have to notice that this implementation ignore spaces
    return len(words[-1])
```

```python
s = "Hello World"
s1 = "   fly me   to   the moon  "
s2='a'
s3=' '
s4='a '

print(lengthOfLastWord(s))
print(lengthOfLastWord(s1))
print(lengthOfLastWord(s2))
print(lengthOfLastWord(s3))
print(lengthOfLastWord(s4))
```
