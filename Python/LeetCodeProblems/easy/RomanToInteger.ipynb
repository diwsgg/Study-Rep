# Roman to Integer

* I can be placed before V (5) and X (10) to make 4 and 9. 
* X can be placed before L (50) and C (100) to make 40 and 90. 
* C can be placed before D (500) and M (1000) to make 400 and 900.

## Given a roman numeral, convert it to an integer.

> * Example 1:

* Input: s = "III"
* Output: 3
* Explanation: III = 3.
> * Example 2:

* Input: s = "LVIII"
* Output: 58
* Explanation: L = 50, V= 5, III = 3.
>* Example 3:

* Input: s = "MCMXCIV"
* Output: 1994
* Explanation: M = 1000, CM = 900, XC = 90 and IV = 4.
 
>* Constraints:

* 1 <= s.length <= 15
* s contains only the characters ('I', 'V', 'X', 'L', 'C', 'D', 'M').
* It is guaranteed that s is a valid roman numeral in the range [1, 3999].

```python
#test 1
def test(s:str)->int:
    num=0

    dict = {
        'I': 1,
        'V': 5,
        'X': 10,
        'L': 50,
        'C': 100,
        'D': 500,
        'M': 1000,
    }
    
    for char in s:
        num += dict[char]
    return num
```

```python
print(test('III'))
print(test('LV'))
```

So we need to check if the current char and the next one is an instance of substracting

```python
#test 2
def test2(s:str):
    num=0

    dict = {
        'I': 1,
        'V': 5,
        'X': 10,
        'L': 50,
        'C': 100,
        'D': 500,
        'M': 1000,
    }
    listConstrains = ['IV','IX','XL','XC','CD','CM']
    
    for i, char in enumerate(s):
        if i!= len(s)-1:
            currentChar = char
            nextchar = s[i+1]
            #CONCATENATE 
            concchar  = currentChar+nextchar
            if concchar in listConstrains:
                num += dict[nextchar]-dict[currentChar]        
                #Problem is that we are iterating again for nextchar, and we need to skip that         
            else:
                num += dict[char]
    return num 
```

```python
print(test2('MCMXCIV'))
```

# Solution

```python
class Solution:
    def romanToInt(self, s: str) -> int:
        num=0

        dict = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000,
        }
        listConstrains = ['IV','IX','XL','XC','CD','CM']
        
        #we will use a while loop, since we can change the index with i+2 to no repeat chars
        i = 0 
        while i < len(s):
            if i<len(s)-1:
                if s[i]+s[i+1] in listConstrains:
                    num += dict[s[i+1]]-dict[s[i]]
                    i+=2
                else:
                    num +=dict[s[i]]
                    i+=1 
            else:
                num +=dict[s[i]]
                i+=1 
        return num


sol = Solution()
```

```python
sol.romanToInt('III')
```

# That was a solution but the best option, the time is N since it will take n*m to check if is on the list, and to check all the string
How we can optimize it 

```python
class Solution:
    def romanToInt(self, s: str) -> int:
        #So the idea is to have the prev value and total
        prev = 0
        total = 0
        hashmap = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }
        #With a foor loop
        for r in s:
            #This will allow us to have a previous value
            #Since we will start the comparations with the "second value"
            if prev > 0:
                #if the previous value is < that the current one, that means is one of the constrains 
                # IV, IX, there are no values like 'VX' or 'CD' in roman numbers so we do not have to worry for this
                if prev < hashmap[r]:
                    total -= prev
                #If the value is equal or greater then we add it 
                else:
                    total += prev
            prev = hashmap[r]
        # Add the last char value, since it rest one, why not rest 
        # If s= IV then total in the loop has total has -1 at the 
        total += hashmap[s[-1]]
        return total
    
sol2 = Solution()
```
