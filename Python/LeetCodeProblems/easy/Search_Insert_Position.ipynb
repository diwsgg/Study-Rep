# Search Insert Position

Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

You must write an algorithm with O(log n) runtime complexity.

> Example 1:

* Input: nums = [1,3,5,6], target = 5
* Output: 2

> Example 2:

* Input: nums = [1,3,5,6], target = 2
* Output: 1

> Example 3:

* Input: nums = [1,3,5,6], target = 7
* Output: 4
 
> Constraints:

* 1 <= nums.length <= 104
*  -104 <= nums[i] <= 104
* nums contains distinct values sorted in ascending order.
* -104 <= target <= 104

```python
nums = [1,3,5,6]
target = 5
```

```python
def searchInsert(nums: list[int], target: int) -> int:
    hashmap = {}
    #this will work, if 
    # the O time should be N, but also if the target is on nums
    # this will not work if the target is not on the list
    for index, num in enumerate(nums):
        if target in hashmap:
            return index
        hashmap[num]=index
    return -1
```

```python
print(searchInsert(nums, target))
```

```python
def searchInsert(nums: list[int], target: int) -> int:
    #binary search works with while loops
    left, right = 0, len(nums)-1
    print(f'value of left:{left} right:{right}')

    while left<=right:

        #get the middle point
        middle = (left+right)//2 #(to get the int value and not float)

        print(f'value of middle {middle}')
        print(f'value of the array on that position {nums[middle]}')

        #Check if we catch the value
        if nums[middle]==target:
            return middle
        #if the value is less than the target we move the left pointer
        elif nums[middle]<target:
            left=middle+1
            print(f'left value: {left}')
        #if is greater the we reduce the right pointer 
        else:
            right=middle-1
            print(f'right value: {right}')

    # so remember this works since the values are sorted in asceding way, that is why we change left when the current 
    # value is less than target, and move right when is greater
    
    #if is not on the array
    return -1
```

# At the end of an unsuccessful search:
> left = right + 1

So:
- **right** points to the last value smaller than the target, or -1 if none exists.
- **left** points to the first position where a value greater than the target begins, or the end of the array.

* We want to insert the target after the smaller values and before the greater values.
That's why:
> return left

works.
* This is closely related to the concept of a lower bound: the first index whose value is greater than or equal to the target.

```python
def searchInsert(self, nums: list[int], target: int) -> int:
    left = 0
    right= len(nums)-1

    #loop
    while left<=right:
        #calculate mid point
        mid = left + (right-left)//2

        if nums[mid]==target:
            #return index of the value
            return mid
        elif nums[mid]<target:
            left = mid+1
        else:
            right = mid-1

    #if no target on the array then we return left
    #since we wanted the first position where a value greater that the target begins
    return left

    #we will return right if we the last value smaller than the target or -1 if non exists
```
