# Two Sum

* You are given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

* You may assume that each input would have exactly one solution, and you may not use the same element twice.

* You can return the answer in any orde

> * Example 1:

* Input: nums = [2,7,11,15], target = 9
* Output: [0,1]
* Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].

> * Example 2:

* Input: nums = [3,2,4], target = 6
* Output: [1,2]
* Example 3:

> * Input: nums = [3,3], target = 6
> * Output: [0,1]
 
> * Constraints:

* 2 <= nums.length <= 104
* -109 <= nums[i] <= 109
* -109 <= target <= 109
* Only one valid answer exists.

> * Follow-up: Can you come up with an algorithm that is less than O(n2) time complexity?

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        #We can rest the first value of the list 
        finalIndixes = []
        #We can know search the element on the list
        #What if the element is not on the list, we can move on the next item and search for the other number
        for index, num in enumerate(nums):
            remainder = target-num
            if remainder in nums[index+1:]:
                #We can add the index of the number that we are working with
                #and look for the index of the remainder if it is on the list
                # But we have to look after the current index 
                finalIndixes.extend([index, nums.index(remainder, index+1)])
                break
        return finalIndixes        
            
sol = Solution()
```

```python
sol.twoSum([2,7,11,15], 9)
```

```python
sol.twoSum([3,2,4], 6)
```

```python
sol.twoSum([3,3], 6)
```

# New version using HashMaps

# WE know that for hashmaps in python we use dicts

we can have a key and a value
* so its like, (number, index)
With this we will take O(1) on search the value of the key and not iterate all the list again

```python
class Solution:
    def twoSum2(self, nums: list[int], target: int) -> list[int]:
        #Empty dict
        allNumbers = {}
        #An array to return
        finalIndixes = []

        for index, num in enumerate(nums):
            remainder = target-num
            #The remainder is not on dict, then we add the current number(key) and the index(value)
            if not remainder in allNumbers:
                allNumbers[num] = index 
            #If we have it then we can return the the indexes,     
            else: 
                finalIndixes.extend([allNumbers[remainder], index])
                #Break the cicle since we do not need to continue
                break

        return finalIndixes   

sol2 = Solution()
```

```python
sol2.twoSum2([2,7,11,15], 9)
```

```python
sol2.twoSum2([3,2,4], 6)
```

```python
sol2.twoSum2([3,3], 6)
```
