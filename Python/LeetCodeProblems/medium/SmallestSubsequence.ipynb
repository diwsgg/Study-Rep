# Get frequency of each letter in a string

```python
#we can use a dict? for fast lookup
def test(s: str)-> str:
    frequency = {}
    #GEt frequency
    for letter in (s):
                            #With get, Return the value for key if key is in the dictionary, else default.
        frequency[letter] = frequency.get(letter,0)+1

    print(frequency)

test("bcabc")
```
