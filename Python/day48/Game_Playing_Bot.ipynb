## If all else fails, we can get use of **XPath**

```python
'''
html structure                  XPath
<body>                --->        /html/body
    <div>             --->        /html/body/div
        <p>           --->        /html/body/div/p
        <ul>          --->        /html/body/div/ul
            <li>      --->        /html/body/div/ul/li
'''
#XPATH can be copied inspecting the element and with that particular option
element_xpath = driver.find_element(By.XPATH, value='//*[@id="site-map"]/div[2]/div/ul/li[3]/a')
```

# Now with that lets do The project

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
#wait until the elements are load it
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import StaleElementReferenceException
#time for the program in general
import time
```

```python
def buyUpgrades(driver, current_product):
    current_cookies = driver.execute_script('return Game.cookies')
    #check the next product
    next_product = current_product + 1
    next_price = driver.execute_script(
        "return Game.ObjectsById[arguments[0]].price",
        next_product
    )

    #move to the next upgrade                                      c
    if current_cookies >= next_price:
        driver.find_element(By.ID, value=f'product{next_product}').click()
        #go for the next upgraded
        current_product+=1
    #if the current cookies are not enough for next product we can buy the current product
    elif current_cookies >= driver.execute_script(f"return Game.ObjectsById[{current_product}].price"):
        driver.find_element(By.ID, value=f'product{current_product}').click()
        #this part is for keep buying the other upgrades and not just keep going for the next one
        # if current_product!=0:
        #     current_product-=1
    else:
        print('does not have enough cookies')
    return current_product   
```

```python
def make_connection():
    chrome_options = webdriver.ChromeOptions()
    chrome_options.add_experimental_option('detach', True)
    
    driver = webdriver.Chrome(options=chrome_options)
    driver.get('https://ozh.github.io/cookieclicker/')
    return driver
```

```python
def main():
    driver = make_connection()
    
    wait = WebDriverWait(
        driver,
        15,
        poll_frequency=0.2,
        ignored_exceptions=(StaleElementReferenceException)
    )

    # Wait until the language button is clickable
    wait.until(
        EC.element_to_be_clickable((By.ID, 'langSelect-ES'))
    ).click()

    current_product = 0

    last_check = time.time()
    maxtime = time.time() + 60 * 1

    while time.time() < maxtime:

        try:
            cookie = wait.until(
                    EC.element_to_be_clickable((By.ID, 'bigCookie'))
            )
            cookie.click()
        except StaleElementReferenceException:
            continue
        #check for the five seconds
        current_time = time.time()
        if current_time - last_check >= 2:
            last_check = current_time
            current_product = buyUpgrades(driver, current_product)

    print('Finished! Check your stats.')
```

```python
main()
```
