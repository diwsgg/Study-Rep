## Use the API from https://pixe.la/

## Adavanced Authentication and POST / PUT / DELETE Request 

```python
import requests
from pathlib import Path
from dotenv import load_dotenv 
import os
```

```python
#As we know get will return the data with some parameters

#But post is the inverse, we will sent the data to the API, and we will receive a code status for that request

#Put is update

#And delate is obviously 
```

```python
file_path = str(Path.cwd().parent)+'/.env'
load_dotenv(file_path)
TOKEN = os.getenv('TOKEN')
USER = (os.getenv('USER_NAME')).lower()
```

```python
#1. step POST
#these are the required parameters
user_params_pixe = {
    'token':TOKEN,
    'username':USER,
    'agreeTermsOfService':'yes',
    'notMinor':'yes',
}
#endpoint
pixela_endpoint = 'https://pixe.la/v1/users'
```

# Create a user for first time (just runt it if we have not an user)

```python
#so instead of using the keyword params, we use, json for POST
response = requests.post(url=pixela_endpoint, json=user_params_pixe)
#This will check if was the answer good
print(response.text)
```

## SO we have created an user for this API, what is next?

## Well before to move on creatring a new graph, we will user **ADVANCED AUTHENTICATION with HTTP**

```python
graph_endpoint = f'{pixela_endpoint}/{USER}/graphs'
#now again do a post request
```

```python
GRAPH_ID = 'graphtest01'
graph_parameters = {
    'id':GRAPH_ID,
    'name':'Weight loss Graph',
    'unit':'kg',
    'type':'float',
    'color':'shibafu',
}
```

```python
#But before call the post service, we need to provide the authentication, in the request header, how we do that?
#HEADER
'''X-USER-TOKEN'''

'''
We can sent the API key on the HEADER for better security
'''

#for that we create a dictionary
header ={
    'X-USER-TOKEN':TOKEN
}
```

```python
#header will not appear on the url, so we are securing our API kEYS LIKE THIS
response_graph = requests.post(url=graph_endpoint, json=graph_parameters, headers=header)

print(response_graph.text)
```

# How do we "create a pixel" on the graph ?

```python
# /v1/users/<username>/graphs/<graphID>
pixel_parameters = {
    'date':'20261019',
    'quantity':'0.5',	
}

pixelCreate_endpoint = f'{pixela_endpoint}/{USER}/graphs/{GRAPH_ID}'
```

```python
response_PIXEL = requests.post(url=pixelCreate_endpoint, json=pixel_parameters, headers=header)

print(response_PIXEL.text)
```

# And thats it, now how do we autofill every day, to change the quantity the date, how?

```python
#First we need to get the date data, for everyday
#Fortunaly we have a method called , strftime, that returns the date in any format that we need

import datetime as dt 

today = dt.datetime.now()
#this method is case sensitive also
today = today.strftime('%Y%m%d')

print(today)
```

```python
#now that we have the date we can updaye the pixel_parameter dictionary
pixel_parameters = {
    'date':today,
    #the quantity is something that can change for us everyday, but we have not to change the date manually now
    'quantity':'0.5',	
}
```

# Now, what about not create a pixel in the graph, maybe update or delete it, so how we do that?

```python
#/v1/users/<username>/graphs/<graphID>/<yyyyMMdd>

#this is optional if the pixel not exists then will create a new pixel with this quantity
pixel_put_parameters = {
    'quantity':'0.75'
}

pixel_Put_endpoint = f'{pixela_endpoint}/{USER}/graphs/{GRAPH_ID}/{today}'
```

# Now we call the method

```python
response_put_pixel = requests.put(url=pixel_Put_endpoint, json=pixel_put_parameters, headers=header)

print(response_put_pixel.text)
```

# To delete a pixel we only just need 
>/v1/users/'<user_name>'/graphs/'<graph_ID>'/'<yyyy_MM_dd> '

and only the header, any parameter needed

```python
#for put and delete is the same endpoint, the differente is we do not pass any json parameter
pixel_Put_endpoint = f'{pixela_endpoint}/{USER}/graphs/{GRAPH_ID}/{today}'

response_delet_pixel = requests.delete(url=pixel_Put_endpoint, headers=header)

print(response_delet_pixel.text)
```
