# Higher or Lower, v1

```python
from d14_Data import data
import d14_Art
import random

keys = ['name','description','country']
```

```python
def newRandomIndex(currentA, newB, oldA):
    # Generate a new B(index for B), excluding A (the previous num2)
    newB = random.choice([i for i in range(len(data))
                               if i!= (currentA and oldA)])
    return newB
```

```python
def printOptions(currentScore, num1, num2):
    #If the score is more than one, we have to
    #Print thre score, move B to A and generate a new index for B but excluding the old index that B has
    if currentScore >= 1:
        print(f"Thats right, Score:{currentScore}")
        # Now we are moving 'B' to 'A', 
        # but we have also to save a temp num1 since we do not wanted the same option for b
        tempnum = num1
        num1 = num2
        #Generate new random choice(index) for B
        num2 = newRandomIndex(num1, num2, tempnum)

    #Print the Options
    print("Compare A: "+", ".join(data[num1][key] for key in keys))
    print(d14_Art.const_vs)
    print("Compare B: "+", ".join(data[num2][key] for key in keys))
    return num1, num2
```

```python
def compareResults(rnum1,rnum2,opt):
    # If a>b and we choose a then its a point.
    # If a>b and we choose b then lose.

    # If b>a and we choose b then its a point.
    # If b>a and we choose a then lose.

    if data[rnum1]["follower_count"]>data[rnum2]["follower_count"]:
        if opt == 'A':
            return True
        else:
            return False
    else:
        if opt == 'B':
            return True
        else:
            return False
```

```python
def main():
    #AT the begining score will be 0 so we do not need to check that everytime
    print("Wanted to play Higher or Lower?")
    another_game = input("YES, NOT").upper()

    while another_game == 'YES':

        continuePlaying = False
        score = 0
        #Generate the two options randomly, but without being the same 
        rnd1, rnd2 = random.sample(range(len(data)),2)

        while not continuePlaying:
            print(d14_Art.logo)

            rnd1, rnd2 = printOptions(score, rnd1, rnd2)
            
            print("Who has the more followers Type A or B")
            answer = input("Type A or B").upper()

            #Temp
            # print(answer)
            # print(f"Compare A: {data[rnd1]["follower_count"]}")
            # print(f"Compare B: {data[rnd2]["follower_count"]}")

            #Function to compare if we accert on the right answer
            checkComparation = compareResults(rnd1,rnd2,answer)

            if not checkComparation:
                print(f"Sorry, that's not right, Total Score: {score}")
                continuePlaying = True
            score+=1

        print("Wanted to play again?")
        another_game = input("YES, NOT").upper()
    print("Have a nice day")
```

```python
main()
```
