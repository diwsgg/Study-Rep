```python
import random
```

```python
def computerChoose(pcScore, totalPcScore, totalUserScore, target):
    #Cicle for pc not have option to stop "adding cards", and just stops for 3 things 
    # 1. Meet target
    # 2. More that target
    # 3. More that totalUserScore less that target 
    while totalPcScore < target:
            pcScore.append(random.randint(1,11))
            totalPcScore += pcScore[len(pcScore)-1]
            if (totalPcScore == target) or (totalPcScore>totalUserScore and totalPcScore<target):
                return compareResults(3)
            elif totalPcScore > target:
                 return compareResults(1)   
```

```python
def compareResults(result):
    if result == 1:
        print("You won")
    elif result == 3: 
        print("You lost")
        #finalData()
    else: 
        return 1
```

```python
def compareUserScore(sumUser, trgt):
    #If its equal we quit the cicle, means user won
    if (sumUser==trgt):
        return 1
    #If not keep going
    elif(sumUser<trgt):
        return 0
    #Means lost
    else:
        return 3
```

```python
def userChoose(ScoreUser, totalScoreUser):
    ScoreUser.append(random.randint(1,11))
    totalScoreUser += ScoreUser[len(ScoreUser)-1]
    return totalScoreUser
```

```python
#get antoher card
def get_Other_CardUser(ScoreUser, totalScoreUser, trgt):
    tempResult = compareUserScore(userChoose(ScoreUser, totalScoreUser),trgt)
    if (tempResult==0):
        return 'y', 0
    if(tempResult == 1 or tempResult == 3):
        return 'n', tempResult         
```

```python
def printingResults(ScoreUser, totalScoreUser, ScorePC, result, totalPCScore):
    if result == 0:
        #Printing the results at the moment
        print(f"Your cards: {ScoreUser} current score {totalScoreUser}")
        print(f"Computer cards {ScorePC}")
    else: 
        print(f"Your final cards: {ScoreUser} and final score {totalScoreUser}")
        print(f"Computer cards {ScorePC} and final score {totalPCScore}")
```

```python
def main():
    #TARGET
    target = 21

    print("Want to play BlackJack")
    optGame = input("yes, no").upper()

    while optGame == 'YES':
        #Variables
        userScore, pcScore = [], []
        #Totals
        totalUserScore, totalPcScore, result = 0, 0, 0    

        #Initial cards
        #User 2 cards, pc 1
        userScore.extend([random.randint(1,11), random.randint(1,11)])
        pcScore.append(random.randint(1,11))

        #Initial sums
        totalUserScore = sum(userScore)
        totalPcScore = pcScore[0]

        #Printing the results at the moment
        printingResults(userScore, totalUserScore, pcScore, result, totalPcScore)

        #Grab another card?
        print("Type 'y' to get another card, Type 'n' to to pass")
        nextRound = input("y,n").lower() 
        while nextRound == 'y' and result == 0:
            nextRound, result = get_Other_CardUser(userScore, totalUserScore, target)
        
        result = compareResults(result)
        if result==1:
            result = computerChoose(pcScore, totalPcScore, totalUserScore, target)    

        if result == 1 or 3:
            print("Wanted to play again?")
            #Clean variables if answer is yes 
            optGame = input("yes, no").upper()

main()
```

```python
#Functions to use in blackjack project
def printingResults(listCardsUser, listCardsPc, ScoreUser, ScorePC, status):
    if status == 'YES':
        print(f"User cards: {listCardsUser}, current score: {ScoreUser}")
        print(f"PC cards: {listCardsPc[0]}")
    else:
        print(f"Your final cards are: {listCardsUser}, final score: {ScoreUser}")
        print(f"PC's final cards: {listCardsPc}, final score: {ScorePC}")

def grabCard(listCards, allcards):
    listCards.append(random.choice(allcards))

def addScore(currentScore, listCards):
    currentScore = sum(listCards)
    return currentScore

def winnerScore(currentScorePC, currentScoreUser, userCards, pcCards):
    if currentScorePC == 21 and (10 in pcCards and 11 in pcCards):
        return "PC"
    elif currentScoreUser == 21 and (10 in userCards and 11 in userCards):
        return "User"
    else:
        return "continue"

def CurrentScore(currentScore, listCards):
    newScore = 0
    if currentScore<=21:
         #still we do need to do changes 
         return listCards, currentScore, "continue" 
    #we now compare if we have a posibility for keepm playing or not
    elif currentScore > 21 and 11 in listCards:
        if not 1 in listCards:
            #"ACE" can change the value from 11 to 1 how we can do this 
            #We need the index for that value change it, do the sum with the new value and come 
            aceList = listCards.copy()
            indexAce = aceList.index(11)
            aceList[indexAce] = 1
            newScore = addScore(newScore,aceList) 
            if newScore>21:
                #lose
                return listCards, currentScore, "PC"
            if newScore<21:
                #can keep playing
                print(f"Ace will be taking as 1 from now, new list: {aceList}, current score {newScore}")
                return aceList, newScore, "continue"
        else: 
            #lose
            print("Can not make another change, for two ICES")
            return listCards, currentScore, "PC"
    else:
        #lose
        return listCards, currentScore, "PC"

def computerGrabs(listCards, currentScore, allcards, result, UserListCards, UserSCORE):
    #WE can do for the first time since computer will grab until user ends
    # And computer will only have one card, we could do this
    grabCard(listCards, allcards)
    #Compare if is a blackjack
    result = winnerScore(currentScore, UserSCORE, UserListCards, listCards)
    if result == 'PC':
        return currentScore, result
    #IF RESULT IS DIFFERENT FROM PC
    else:
        while currentScore<16:
            grabCard(listCards, allcards)
            currentScore = addScore(currentScore, listCards)
        return currentScore, result

def FinalResults(userScore, pcScore):
    if userScore == pcScore:
        return "even"
    elif userScore>pcScore:
        return"User"
    else:
        return "PC"
```

```python
#Add a list for the cards
availablecards = [11, 2, 3, 4, 5, 6, 7, 8, 9, 10, 10, 10, 10]

dictOptions = {
    "PC": "You lose",
    "User": "You win",
    "even": 'Its a draw',
}
anotherGame = ''
# cicle to repeat another game 
print("Welcome to BlackJack")
anotherGame = input("Wanted to play?, [yes no]\n:").upper()
# -----------------------------
while anotherGame == 'YES':
    user_cards, pc_cards = [], []
    user_score, pc_score = 0,0

    #Again first we put two cards on user_cards and 1 on pc_cards
    user_cards.extend([random.choice(availablecards),random.choice(availablecards)]) #WE can call the function twice but we can dop this also
    grabCard(pc_cards, availablecards)

    #After this point another cicle since it was given the initial cards
    quit = 'YES'
    # ----------------------------------------------------
    while quit == 'YES':
        #Sums on scores
        user_score = addScore(user_score,user_cards) 
        pc_score = addScore(pc_score, pc_cards)

        #Print the cards and score
        printingResults(user_cards,pc_cards, user_score, pc_score, quit)

        #We compare if the user or computer has blackjack (21) (ace + 10, tis  blacjack not just hit 21)
        resultWinner = winnerScore(pc_score, user_score, user_cards, pc_cards)

        if resultWinner =="continue":
            #Now we only compare if the user_score is >21
            user_cards, user_score, resultWinner = CurrentScore(user_score, user_cards)
        
        if resultWinner == "PC":
            print(dictOptions[resultWinner])
            quit='NO'
            break
        elif resultWinner == 'User':
            print(dictOptions[resultWinner])
            quit='NO'
            break
        else:
            print("Want to grab another card?: ")
            quit = input("YES, NO\n:").upper()
            if quit == 'YES':
                grabCard(user_cards, availablecards)

    #If we have lose we do not need to continue grabing cards for computer, 
    if resultWinner == 'continue':
        #Our cicle for user works, even if he hits 21 without a blackjack we will not win until the end
        pc_score, resultWinner = computerGrabs(pc_cards, pc_score, availablecards, resultWinner, user_cards, user_score)

        #Print the cards and score
        printingResults(user_cards,pc_cards, user_score, pc_score, quit)
        #If pc has no blackjack
        if resultWinner == 'continue':
            #Check if computer has gone over 21
            if pc_score>21:
                resultWinner = 'User'
                print(dictOptions[resultWinner])
            else:
                #User lose, but computer did not hit a blackjack
                resultWinner = FinalResults(user_score, pc_score)
                print(dictOptions[resultWinner])
        else: 
            print("PC HITS A BLACKJACK, sorry you lose")
    print("-"*30)
    print("Want to play again?")
    print("-"*30)
    anotherGame = input("YES NO\n:").upper()
print("Have a nice day")
```
