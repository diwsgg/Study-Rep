```python
import random
```

```python
def compareResults(result):
    if result == 1:
        print("You won")
        #finalData()
    elif result == 3: 
        print("You lost")
        #finalData()
    else: 
        return 1
```

```python
def finalData():
    global userScore, pcScore, totalUserScore, totalPcScore

    print(f"Your cards: {userScore} yor final score {totalUserScore}")
    print(f"Computer cards {pcScore}, final score {totalPcScore}")
```

```python
def printingResults(ScoreUser, totalScoreUser, ScorePC):
    #Printing the results at the moment
    print(f"Your cards: {ScoreUser} current score {totalScoreUser}")
    print(f"Computer cards {ScorePC}")
```

```python
#get antoher card
def get_Other_CardUser(ScoreUser, totalScoreUser, trgt):
    tempResult = compareUserScore(userChoose(ScoreUser, totalScoreUser),trgt)
    if (tempResult==2):
        return 'y', 0
    if(tempResult == 1 or tempResult == 3):
        return 'n', tempResult         
```

```python
def userChoose(ScoreUser, totalScoreUser):
    ScoreUser.append(random.randint(1,11))
    totalScoreUser += ScoreUser[len(ScoreUser)-1]
    return totalScoreUser
```

```python
def compareUserScore(sumUser, trgt):
    #If its equal we quit the cicle
    if (sumUser==trgt):
        return 1
    #If not keep going
    elif(sumUser<trgt):
        return 0
    #Means lost
    else:
        return 3
```

```python
def computerChoose(pcScore, totalPcScore, totalUserScore, target):
    #Cicle for pc not have option to stop "adding cards", and just stops for 3 things 
    # 1. Meet target
    # 2. More that target
    # 3. More that totalUserScore less that target 
    while totalPcScore < target:
            pcScore.append(random.randint(1,11))
            totalPcScore += pcScore[len(pcScore)-1]
            if (totalPcScore == target) or (totalPcScore>totalUserScore and totalPcScore<target):
                return compareResults(3)
            elif totalPcScore > target:
                 return compareResults(1)   
```

```python
def main():
    #TARGET
    target = 21

    print("Want to play BlackJack")
    optGame = input("yes, no").upper()

    while optGame == 'YES':
        #Variables
        userScore, pcScore = [], []
        #Totals
        totalUserScore, totalPcScore, result = 0, 0, 0    

        #Initial cards
        #User 2 cards, pc 1
        userScore.extend([random.randint(1,11), random.randint(1,11)])
        pcScore.append(random.randint(1,11))

        #Initial sums
        totalUserScore = sum(userScore)
        totalPcScore = pcScore[0]

        #Printing the results at the moment
        printingResults(userScore, totalUserScore, pcScore)

        #Grab another card?
        print("Type 'y' to get another card, Type 'n' to to pass")
        nextRound = input("y,n").lower() 
        while nextRound == 'y' and result == 0:
            nextRound, result = get_Other_CardUser(userScore, totalUserScore, target)
        
        result = compareResults(result)
        if result==1:
            result = computerChoose(pcScore, totalPcScore, totalUserScore, target)    

        if result == 1 or 3:
            print("Wanted to play again?")
            #Clean variables if answer is yes 
            optGame = input("yes, no").upper()

main()
```
