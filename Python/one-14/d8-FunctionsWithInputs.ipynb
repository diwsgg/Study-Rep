```python
def calculate_love_score(name1, name2):

    #Check for "TRUE" and "LOVE" how many times the letters of the names appears 
    namesTogether = (name1+name2).upper()

    arrayTrue = ['T','R','U','E']
    arrayLove = ['L','O','V','E']
    
    totalTrue = 0
    totalLove = 0

    for letter in namesTogether:
        if letter in arrayTrue:
            totalTrue+=1
        if letter in arrayLove:
            totalLove+=1
    #WE add those counts together, not as a sum but as strings
    #Problem the loveScore is a string so after convert them to str we need to convert it to an int 
    loveScore = int(str(totalTrue)+str(totalLove))
    
    print(loveScore)


calculate_love_score("Kanye West", "Kim Kardashian")
```

# Caesar Cipher

```python
alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

direction = input("Type 'encode' to encrypt, type 'decode' to decrypt: ").lower()
text = input("Tyoe your message: ").lower()
shift = int(input("Type the shift number: "))

#TODO-1, Create a function called encrypt() that takes the original_txt and shift_amount as 2 inputs
#TODO-2, Inside the 'encrypt' function, shift each letter of the original_text forwards in the alphabet by the shift_amount and print the encrypted text
#TODO-4, What happens if you try to shift Z forwards by 9? How we can fix that??
#TODO-3, Call the 'encrypt' functiopn and pass in the user inputs.

#Optional to see whats the values 
print(direction)
print(text)
print(shift)
```

```python
#TODO-1
def encrypt(original_text, shift_amount):
    #TODO-2
    #We can use the function index (to know the position of an item in a list) 
    #
    text_encode = ''

    for letter in original_text:
        if letter !=' ': 
            #We need to know the index of the letter that we pass in our text into our alphabet
            index_letter = alphabet.index(letter)
            # Now that we know the index and we add the shift we can replace the letter from that new position
            # We are not replacing the letter but creating a new string 

            #TODO-4
            '''
            This works but what if the letter is a Z and shift is 3, we do not have more indexes in the array how we can solve this 

            text_encode += alphabet[index_letter+shift_amount]

            This can be fix with using the operator '%' and lenght of the array that is '26' for make it a cicle between 0 and 25, but if we do not know what the lenght is we can use %len(array)
            '''
            
            text_encode += alphabet[(index_letter+shift_amount)%len(alphabet)]
        else:
            text_encode += ' '
    
    print(f"Original word:{original_text}\nEncrypted Word: {text_encode}")
```

```python
#TODO-3
encrypt("celular", 200)
```

# Now we can continue to make the decryption

```python
def decrypt(encrypt_text, shift_amount):
    text_encode = ''
    for letter in encrypt_text:
        if letter !=' ': 
            index_letter = alphabet.index(letter) 
            #We can change the logic
            # Instaed of adding the shift amount, now we can rest it           
            text_encode += alphabet[(index_letter-shift_amount)%len(alphabet)]
        else:
            text_encode += ' '
    
    print(f"Original word:{encrypt_text}\nEncrypted Word: {text_encode}")
```

```python

decrypt("uwdmdsj", 200)
```

# Now we need to combine encrypt and decrypt function into a single function called caesar()
* Use the value of the user chosen "direction" variable to determine which functionality to use
* Call the caesar function instaed of encrypt/decrypt and pass in all three variables
  * direction/text/shift

```python
def caesar(direction, text, shift):
    if direction != 'encrypt' and direction != 'decrypt':
        print("Invalid option")
    elif direction == 'encrypt':
        encrypt(text,shift)
    else: 
        decrypt(text,shift)
```

```python
#TODO-1
def caesar2(direction, text, shift_amount):

    #Check if the "direction" is a valid option before starting
    if direction != 'encrypt' and direction != 'decrypt':
        print("Invalid option")
        return
    #if WE ARE decrypting then we do not need to compare everytime inside the cicle just one, and we can change the value of
    #shift amount to make it negative and with that we are subtraction and not adding  
    elif direction == 'decrypt':
            shift_amount *=-1
        
    #Continue normal
    text_encode = ''
    for letter in text:
        if letter != ' ': 
            index_letter = alphabet.index(letter)
            text_encode += alphabet[(index_letter+shift_amount)%len(alphabet)]
        else:
            text_encode += ' '
    
    print(f"Word: {text}\n{direction.upper()} Word: {text_encode}")
```
