### But we do not wanted 5seconds we want 25 min so how do we do that

> So this is basically
* 1 second * 60 seconds = 1 min
* 2 second * 60 seconds = 2 min
* ...
* 25 seconds * 60 seconds = 25 min
> But this will be seen as 300 seconds, 245 seconds, 1500 seconds
> * We need to convert this to 25:00 

* 245//60 = 4 min, 
> for the seconds
* 245%60 = remainder (seconds)

```python
window = tk.Tk()
counter = 10 #i wanted to convert this to min
window.title("Pomodoro")
window.config(padx=100, pady=100, bg=YELLOW)
 
canvas = tk.Canvas(width=200, height=250, bg=YELLOW, highlightthickness=0)
tomato_img = tk.PhotoImage(file=tomato_path)
canvas.create_image(100, 112, image=tomato_img)
timer_text = canvas.create_text(100, 130, text='00:00', fill='purple', font=(FONT_NAME, 24, 'bold'))
canvas.grid(column=1, row=1)

#buttonclicked events
def button_start_click():
    #We calculate the min and seconds 
    total = counter*60
    minutes = total//60
    seconds = total%60
    #and convert the seconds to 59 and rest minutes -1 if the seconds are 0
    if seconds==0:
        minutes-=1
        seconds=59
    #for last we pass this
    countdown(minutes, seconds)

#The method has to be after we create the the canvas obj
def countdown(minutes, seconds):
    canvas.itemconfig(timer_text, text=f"{minutes}:{seconds}")
    #first check seconds
    if seconds>0:
        window.after(1000, countdown,minutes, seconds-1)
    #then if we have minutes
    elif minutes>1 and seconds==0:
        window.after(1000, countdown,minutes-1, seconds=59)
    #if both are 0
    elif minutes==0 and seconds==0:
        print("countdown finish")
        
#start button
start_button = tk.Button(text="Start",
    font=(FONT_NAME, 15, "bold"),
    bg='white',
    command=button_start_click
)
start_button.grid(column=0, row=2)

window.mainloop()
```

```python
window = tk.Tk()
#i wanted to convert this to min so instead of writing 10 i will do 
#10*60
counter = 600 
window.title("Pomodoro")
window.config(padx=100, pady=100, bg=YELLOW)
 
canvas = tk.Canvas(width=200, height=250, bg=YELLOW, highlightthickness=0)
tomato_img = tk.PhotoImage(file=tomato_path)
canvas.create_image(100, 112, image=tomato_img)
timer_text = canvas.create_text(100, 130, text='00:00', fill='purple', font=(FONT_NAME, 24, 'bold'))
canvas.grid(column=1, row=1)

#buttonclicked events
def button_start_click():
    countdown(counter)

#Instead of doing to ouch elif we are gonna change the logic and use dynamic typing
def countdown(counter):
    #We are doing the maths for the counter every loop
    minutes = counter//60
    #or using math.floor(counter/60) is the same 
    
    #In the next cicle we are creating seconds as int
    seconds = counter%60

    #Now the dinamic part is this
    if seconds == 0:
        #We will convert the seconds (int) to an string variable
        seconds = '00'
    elif seconds<10:
        # seconds = '0'+str(seconds)
        seconds = f'0{seconds}'

    canvas.itemconfig(timer_text, text=f"{minutes}:{seconds}")
    if counter>0:
        window.after(1000, countdown,counter-1)

#start button
start_button = tk.Button(text="Start",
    font=(FONT_NAME, 15, "bold"),
    bg='white',
    command=button_start_click
)
start_button.grid(column=0, row=2)

window.mainloop()
```

#### There are some problems that I notice
* The start button can be press many times as the user want and that will cause a problem
* The reset buton as has a None value the first time if is press before a cicle then it will produce an error

## Here is the solution

```python

# ---------------------------- CONSTANTS ------------------------------- #
PINK = "#e2979c"
RED = "#e7305b"
GREEN = "#9bdeac"
YELLOW = "#f7f5dd"
FONT_NAME = "Courier"
WORK_MIN = 1
SHORT_BREAK_MIN = 2
LONG_BREAK_MIN = 3
reps = 0
timer = None

window = tk.Tk()

window.title("Pomodoro")
window.config(padx=100, pady=100, bg=YELLOW)
 
canvas = tk.Canvas(width=200, height=250, bg=YELLOW, highlightthickness=0)
tomato_img = tk.PhotoImage(file=tomato_path)
canvas.create_image(100, 112, image=tomato_img)
timer_text = canvas.create_text(100, 130, text='00:00', fill='purple', font=(FONT_NAME, 24, 'bold'))
canvas.grid(column=1, row=1)

#labels
#timer-label
timer_label = tk.Label(text="Timer",
    font=(FONT_NAME, 50, "bold"), 
    bg=YELLOW, fg=GREEN, highlightthickness=0
)
timer_label.grid(column=1, row=0)

#checkmark-label
checkmark_label = tk.Label(
    font=("Arial", 50, "bold"), 
    bg=YELLOW, fg=GREEN, highlightthickness=0
)
checkmark_label.grid(column=1, row=3)

#buttonclicked events
def button_start_click():
    #we disable the button
    start_button.config(state='disabled')
    global reps
    reps+=1
    work_sec = WORK_MIN * 60
    short_break_sec = SHORT_BREAK_MIN *60
    long_break_sec = LONG_BREAK_MIN *60

    if reps==8:
        timer_label.config(text='Long break time', fg= RED)
        countdown(long_break_sec)
        reps=0
    elif reps%2==0:
        timer_label.config(text='break time', fg= PINK)
        countdown(short_break_sec)
    else:
        timer_label.config(text='Work time', fg= GREEN)
        countdown(work_sec)

def button_end_click():
    global reps, timer
    # This will fix the error when timer is None 
    # and the user press the reset button before the start button 
    if timer is not None:
        window.after_cancel(timer)
        timer = None
    canvas.itemconfig(timer_text, text='00:00')
    #labels
    timer_label.config(text='Timer')
    checkmark_label.config(text='')
    reps=0
    #we activate the start button again
    start_button.config(state="normal")

def countdown(counter):
    minutes = counter//60
    seconds = counter%60

    if seconds == 0:
        seconds = '00'
    elif seconds<10:
        seconds = f'0{seconds}'

    canvas.itemconfig(timer_text, text=f"{minutes}:{seconds}")
    if counter>0:
        global timer
        timer = window.after(1000, countdown,counter-1)
    else:
        button_start_click()
        mark = ''
        worksession = reps//2
        for _ in range(worksession):
            mark+='✔'
        checkmark_label.config(text=mark)

#buttons
#start button
start_button = tk.Button(text="Start",
    font=(FONT_NAME, 15, "bold"),
    bg='white',
    command=button_start_click
)
start_button.grid(column=0, row=2)

#reset button
reset_button = tk.Button(text="Reset",
    font=(FONT_NAME, 15, "bold"),
    bg='white', 
    command=button_end_click
)
reset_button.grid(column=2, row=2)


window.mainloop()
```
