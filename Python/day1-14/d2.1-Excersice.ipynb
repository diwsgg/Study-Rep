## TIP CALCULATOR

**WE ARE GOINT TO CREATE A CALCULATOR, FOR TIPS**
* WE CAN PUT THE AMOUNT OF MONEY WE SPENT
* The amount of people that is sharing the account
* And the porcentage of the tip

```python
print("Welcome to the TIP calculator");
```

# Asked the information to the user

```python
subtotal_bill = input("What was the total bill? $")
tip_amount = input("How much tip would you like to give? 10%, 12%, 15%? ")
total_people = input("How many people to split the bill? ")
```

```python
print(type(subtotal_bill))
print(type(tip_amount))
print(type(total_people))
#Even that is all numbers, the input function will detected as strings 
print(subtotal_bill)
print(tip_amount)
print(total_people)
```

# Now we do the operations

```python
#Get the amount of money of the porcentage
porcentage = ((int(tip_amount)*int(subtotal_bill))/100)
#final total
total_bill = ((float(subtotal_bill)+float(porcentage))/int(total_people))
```

# Show the total of each person

```python
print(f"Each person should pay: ${round(total_bill,2)}")
```

# To not have to convert every time the variables in the operations
If we know that we expect an integer or a float, some other that a string we could do this

```python
subtotal_bill = float(input("What was the total bill? $"))
tip_amount = float(input("How much tip would you like to give? 10%, 12%, 15%? "))
total_people = int(input("How many people to split the bill? "))
```

> Now we do not have to put the type for convert those variables

```python
#Get the amount of money of the porcentage
porcentage = ((tip_amount*subtotal_bill)/100)
#final total
total_bill2 = round((subtotal_bill+porcentage)/total_people,2)
```

> And the result is the same

```python
print(f"Each person should pay: ${total_bill2}")
```
