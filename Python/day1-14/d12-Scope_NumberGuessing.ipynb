```python
#Like this?
print(game(drink_potion()))
```

```python
#WEll actually we need to modify the function to be accessible

# A nested function how we can call them

def game():
    def drink_potion():
        potion_strength = 2
        print(potion_strength)
    drink_potion()

#And now we can only just call game()
print(game())
```

Python does not have block scope
* The variables created nested in other blocks of code (loops, if...), do not get local scope
* They are given function scope if they are within a function or global scope if they are not.

## Prime Number Checker
### Prime numbers are numbers that can only be cleanly divided by themselves and 1. Wikipedia  

* You need to write a function called is_prime() that checks whether if the number passed into it is a prime number or not.  It should return True or False.
- e.g.

* 7 is a primer number because it is only divisible by 1 and itself.
* But 4 is not a prime number because you can divide it by 1, 2 or 4.
* NOTE: 2 is a prime number because it's only divisible by 1 and itself, but 1 is not a prime number because it is only divisible by 1.
----------
----------
> * Example Input 1
>* 73
>* Example Output 1
>* True
----------
----------
>* Example Input 2
>* 75
>* Example Output 2
>* False

```python
def is_prime(num):
    if num == 1:
        return False
    #We need to iterate from 2 to num and if one of the iterations ==0 then is not a prime
    #since all the primes will be a remainder of x.123134654654..... or 1 but not 0
    #Instaed a not prime could have a remainder of 0
    for i in range(2, num):
        if num%i == 0:
            #IS not a prime, it has anbother diviser that give us 0
            return False
    #HAS no divisers but 1 and itself so is a prime
    return True

is_prime(73)
```

# Number guessing

#### We can choose between 2 difficulties
* hard 5 lives
* easy 10 lives

Need to guess a number between 1-100
* if the guess is < random number
  * print(to loow)
* if guess is > random number
  * print(to high)
* else:
  * print(that's the number)

- Also for the program if at the end the user has no lives and has not guessed it, then we can print the random number

```python
import random
```

```python
def compareGuessNumber(guessNumber, lives):
    if guessNumber == numberChoice:
        print("Thats the number")
        return 0, lives-1
    elif guessNumber>numberChoice:
        print("Too high\n")
        return 1, lives-1
    elif guessNumber<numberChoice:
        print("Too low\n")
        return 1, lives-1
```

```python
another_game = 'YES'
while another_game == 'YES':
    print("Welcome to the Number Guessing Game\n I'm thinking of a number between 1-100\n Choose a difficulty: ")
    mode_level = input("easy, hard").lower()

    numberChoice = random.randint(1,100)
    guessNumber = 0
    flag = False
    #Hint
    # print(f"the number is {numberChoice"})
    if mode_level == "easy":
        lives = 10
    else:
        lives = 5
    #Cicle from here
    while not flag and lives>0: 
        print(f"You have {lives} attemps, remainding to guess the number\nMake a guess: ")
        guessNumber = int(input(" "))
        print(f"Your number was {guessNumber}")
        #CALL THE FUNCTION
        result, lives = compareGuessNumber(guessNumber, lives)

        if result == 0:
            flag = True
        if lives == 0:
            print(f"The number was {numberChoice}, better luck next time")
    print("Wanted to play another game: ")
    another_game = input("yes, no").upper()

print("Have a nice day")
```
