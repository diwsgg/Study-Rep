# Challenge

## Leap Year
#### * 💪 This is a difficult challenge! 💪 

* Write a program that returns True or False whether if a given year is a leap year.

* A normal year has 365 days, leap years have 366, with an extra day in February. The reason why we have leap years is really fascinating, [video](https://www.youtube.com/watch?v=xX96xng7sAE)

* This is how you work out whether if a particular year is a leap year. 
> - on every year that is divisible by 4 with no remainder
> - except every year that is evenly divisible by 100 with no remainder 
> - unless the year is also divisible by 400 with no remainder   

* If English is not your first language, or if the above logic is confusing, try using this flow chart.

>* e.g. The year 2000: 

>* 2000 ÷ 4 = 500 (Leap)  
>* 2000 ÷ 100 = 20 (Not Leap)  
>* 2000 ÷ 400 = 5 (Leap!)  
* So the year 2000 is a leap year. 

* But the year 2100 is not a leap year because: 

>* 2100 ÷ 4 = 525 (Leap)  
>* 2100 ÷ 100 = 21 (Not Leap)  
>* 2100 ÷ 400 = 5.25 (Not Leap)  

***Warning***

> Example Input 1

> Example Input 2

```python
#Leap year

def is_leap_year(year):
    if year % 4 == 0 and year % 100 == 0 and year % 400 == 0:
        return True
    return False

#The error will occur with 2020 or 2024, since for 100 and 400 is different to 0 but 
# at the same time 2020 is a leap year but how this is happening, 
# 2020%4=0 --> true
# 2020%100=20 --> False for remainder = 0  ---- WE have to see this
# Since the condition says "except if is divisible by 100 and remainder is 0"
# This is not true ion this case, so its a year leap, and we do not have to check for "mod 400"   
is_leap_year(2020)
```

# Solution

```python
def is_leap_year(year):
    #Nested if 

    if year % 4 == 0:
        #true
        if year % 100 == 0:
            #If is divisible we keep on to verify the last condition to check if is a leap year or not
            if year % 400 == 0:
                return True
            else:
                return False
        #If year%100!=0 that means its not the exception to not be a leap year
        else: 
            return True        
    else:
        return False

is_leap_year(1987)
```

# Calculator

### WE need to make a calculator with the operations
* "+"
* "-"
* "*"
* "/"
* For **TWO** nummbers
A cicle where it stops when the user decides to that's it

```python
#Operations fucntions
def add(n1, n2):
  return n1 + n2
 
def subtract(n1, n2):
  return n1 - n2
 
def multiply(n1, n2):
  return n1 * n2
 
def divide(n1, n2):
  if n2==0:
    return "Number 2 need to be greater that number 1, division not possible"
  return n1 / n2
```

```python
#Final function
def calculator(option, a,b):
    #switch in python
    match option:
        case 1:
            return add(a,b)
        case 2:
            return subtract(a,b)
        case 3:
            return multiply(a,b)
        case 4:
            return divide(a,b)
        #Case where the data did not match any case
        case _:
            return "Invalid option"
```

```python
def printOptions(dictOperations):
    for key in dictOperations:
        print(key)
```

```python
#main

def main():
    ContinueFLag = False
    dictOperations = {
        '+': 1,
        '-': 2,
        '*': 3,
        '/': 4,
        } 
    while not ContinueFLag:
        print("Welcome to calculator app")
        print(f"What operation do you want to do:")
        '''
        For print without a FOR LOOP the keys in a dictionary without the values we can use

        list(dictOperations.keys()) --> ['key1','key2','key3',]
        or
        dictOperations.keys()  --> nameDictionary(['key1','key2','key3',])

        This could work if we do not care about how it looks, but if we prefer we can use a for loop
        '''
        printOptions(dictOperations)
        print("*"*50)
        option = input("Operation: ")
        # while not dictOperations[option] in dictOperations:
        #          option = input("Invalid operation: ")

        number1 = int(input("Digit number 1: "))
        number2 = int(input("Digit number 2: "))

        print(f"The numbers are: {number1} and {number2}")
        print("*"*50)

        result = calculator(dictOperations[option], number1, number2)
        print("*"*50)

        print(f"The result of the operation {option} is: {result}") 
        print("*"*50)
        print("*"*50)

        #Asked if wanted another operation
        nextCIcle = input("Wanted to do another operation: [YES, NO]").upper()
        print("-*-"*50)

        if nextCIcle == 'NO':
            ContinueFLag = True
            print("Thanks good bye")

main()  
```

```python
#Operations fucntions
def add(n1, n2):
  return n1 + n2
 
def subtract(n1, n2):
  return n1 - n2
 
def multiply(n1, n2):
  return n1 * n2
 
def divide(n1, n2):
  if n2==0:
    return "Number 2 need to be greater that number 1, division not possible"
  return n1 / n2

dictOpp = {
  '+': add,
  '-': subtract,
  '*': multiply,
  '/': divide,
}
```
