# FUNCTIONS, CODE BLOCKS, and While LOOPS

# Functions
Simplest form is just a wrapper name for a block of code.
* You give it a name and then when you call the function by that name, all the code within the function block will be executed.


```python
def my_function():
    print("Hello")
    print("Bye")
```

```python
#Now we can call the function as
my_function()
#As we do not pass any variable, we can call it like this
```

Of course we can call functions inside a function to call another function like:

```python
def function1():
    function2()
    function3()
    #...... 
    # Even itself
    #This will create an infinite call, we need to be carefull with this 
    #function1() 


def function2():
    print("Hello")

def function3():
    print("bye")

function1()

#If we wanted the function to be print N times if we need it we can use a loop like
for _ in range(2):
    print(f"Loop number:{_}")
    function1()
```

# While Loops

```python
def function1():
    function2()
    function3()
    #...... 
    # Even itself
    #This will create an infinite call, we need to be carefull with this 
    #function1() 


def function2():
    print("Hello")

def function3():
    print("bye")

#We are going to print the function but now with a while loop
number_iterations = 2
#This will print the function1() twice as the for above 
while number_iterations > 0:
    function1()
    number_iterations-=1
    print(number_iterations)
```

The difference for the use between a 
* for loop
* while loop

IS that, 
* For loops, we need them for iterate each element from a list, or a range specific
* While loop, we need them for do something until reaches a flag
