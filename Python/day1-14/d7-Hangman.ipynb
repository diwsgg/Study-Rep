# The problem is that the array where we store the guess word until we reach the word, is not stored only append every cicle, so we need another way

```python
chosen_word = random.choice(word_list)
print(chosen_word)

flag = False
#We need a list for store the results
#This will be the change since it is not inside the cicle we will not clean the variable and we can store all the correct options
correct_letters = []
while not flag:
        guess = input("Guess ONE letter: ").lower()
        
        #Variable str
        placeholder = "" 

        for letter in chosen_word:
                '''
                If its on the array "correct_letters", then we can print the letters on placeholder
                Lets suppused "guess = e" then "correct_letters=['a', 'l', 'e']" we have the letter on the array
                we do not need to added to the array, just print "placeholder" until we reached letter = e (the end),
                The result: "a__le" 
                '''
                if letter in correct_letters:
                        placeholder += letter

                # If the user has guess "a" and the word is "apple" then we can print "a____", and added to the array
                elif letter == guess:
                        placeholder += letter
                        correct_letters.append(letter)
                else:
                        placeholder += "_"

        print(placeholder)
        #Check if we have _ in the placeholder
        if "_" not in placeholder:
                flag=True
                print("YOU WON")
    
```

And also print the stages of the lives that we are

```python
stages = [
'''
 +--+
 |  |
 o  |
/|\ |
/ \ |
''',
'''
 +--+
 |  |
 o  |
/|\ |
/   |
''',
'''
 +--+
 |  |
 o  |
/|\ |
    |
''',
'''
 +--+
 |  |
 o  |
/|  |
    |
''',
'''
 +--+
 |  |
 o  |
/   |
    |
''',
'''
 +--+
 |  |
 o  |
    |
    |
''',
'''
 +--+
 |  |
    |
    |
    |
'''
]
```

```python
#We now need to repeat until it reaches the guessword on the list or the lives gets 0

chosen_word = random.choice(word_list)
#print(chosen_word)

flag = False
flLives = False
lives = 6
#We need a list for store the results
correct_letters = []
while not flag and lives > 0:
        guess = input("Guess ONE letter: ").lower()
        
        #Variable str
        placeholder = "" 
        #Check if it is on chosen_word
        for letter in chosen_word:
                if letter in correct_letters:
                        placeholder += letter
                elif letter == guess:
                        flLives = True
                        placeholder += letter
                        correct_letters.append(letter)
                else:
                        placeholder += "_"

        print(placeholder)
        #Check if we have _ in the placeholder
        if "_" not in placeholder:
                flag=True
                print("YOU WON")
                break
        if not flLives:
                lives-=1
                if lives==0:
                        print("GAME OVER")
                        break
        #Print the stages of the live and how many we have left        
        print(stages[lives])
        print (f"You have {lives} lives left") 
        print("-"*50+"\n")
        #Re-assign the flag for lives
        flLives = False
    
```
