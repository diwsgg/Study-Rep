## List comprehension

### In a normal way, if we have an initial array, and then we wanted aonther array but with the numbers +1, we can do it like this:

```python
numbers = [1,2,3]
new_list = []

#Create the new list but adding one to each element
for n in numbers:
    add_1 = n+1
    new_list.append(add_1)

print(new_list)
```

### With the list comprehension we can do this more simple

```python
numbers = [1,2,3]

# WE are adding 1 at each number of the initial array, 
# and creating a new array at the same time
# without modifying the initial array
new_list = [i+1 for i in numbers]

print(new_list)
```

## First we will get the content of the files

```python
from pathlib import Path

folder_path = Path.cwd()

file1_path = folder_path / "files" / "file1.txt"
file2_path = folder_path / "files" / "file2.txt"

#Now have the content of the files
with open(file1_path, 'r') as file1:
    #Create a comprehansion list, for ignore the \n spaces
    file_one = [n.rstrip() for n in file1.readlines()]

with open(file2_path, 'r') as file1:
    file_two = [n.rstrip() for n in file1.readlines()]

#It is important to mention that this will return a string
# not int 
print(file_one)
print(file_two)

```

### If I wanted an int i have to convert it but i can do it from the beginning and as the problem is asking us to have the output as int

```python
with open(file1_path, 'r') as file1:
    #Create a comprehansion list, for ignore the \n spaces
    file_one = [int(n.rstrip()) for n in file1.readlines()]

with open(file2_path, 'r') as file1:
    file_two = [int(n.rstrip()) for n in file1.readlines()]

print(file_one)
print(file_two)
```

## Now we will create another comprehension list that havem only the elements that are in both arrays

```python
# Create a list with the numbers of any of the two arrays, and 
# with the if statement we ask, newN (the element) is in the other array?

result1 = [newN for newN in file_two if (newN in file_one)] 

print(result1)
```
