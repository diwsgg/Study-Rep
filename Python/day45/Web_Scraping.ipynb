# Find and Select Particular Elements

```python
#a function call find_all() is the one we will use to print all the tags (it will return a list)
list_all_paragrahps = soup.find_all(name='p')
print(list_all_paragrahps)
#the first value
print(list_all_paragrahps[0])
print(type(list_all_paragrahps[0]))
```

```python
#what if I only the text and not the tags?
#with the information that we received we are going to make a for loop to get only the text for each tag

for tag in list_all_paragrahps:
    #and use a method called get text
    #this could be done if we wanted a list with only the text of each paragrah or whatever tag
    print(tag.get_text())
```

# As we see we can search now for all the tags, but what if I am only looking for a specific one, how?

```python
# with the find method we will return the first item that match the query
# so we know that we add the name of the tag, and we have **kwargs so we can add id as variable
heading = soup.find(name='h1', id='name')

print(heading)
```

```python
#but class is a reserved word so how we use it?
# heading = soup.find(name='h3', class='heading')

#to fix this is so simple adding a _ to the word class
heading = soup.find(name='h3', class_='heading')

print(heading)
```

```python
span_tag_text_link = NewSoup.select("span.titleline > a")

#get text 
titles = [a.get_text() for a in span_tag_text_link]
#get link
links = [a.get(key='href') for a in span_tag_text_link]

print(titles)
print(links)
#check if have the same amount of data
print(len(titles))
print(len(links))
```

## Get the votes for each one
#### We are going to implement the solution with **select** and **find_all** to see the diferrence

```python
scores = [a.get_text() for a in NewSoup.select(selector='span.score')]
print(f"Scores with select method: {scores}")
print(len(scores))

scores = [a.get_text() for a in NewSoup.find_all(name='span', class_='score')]
print(f"Scores with find_all method: {scores}")
print(len(scores))
```

### And is the same, the only difference is how we put our selectors

## Can we combine all in the same array
# **ZIP method**

Python has a function called zip, and in this case we will use
# **zip with list comprehension**

Important things to know about zip
* **zip()** stops when the shortest list ends.

* **zip()** helps us to iterate each value one at time of the different iterables
    * Like tuples (title, link, scores) and unpacks each tuple to the variables

* **zip()** produces each group as needed instead of creating a complete list of tuples in memory.

* **zip()** matches elements by position, not by their content. 
    * So ours lists must be correctly aligned for each article to receive its own link and vote count.

```python
articles = [
    #{}we encapsulate each value in a dictionary
    #first the key, then the value
    {"title": title, "link": link, "votes": vote}
    for title, link, vote in zip(titles, links, scores)
]
```

```python
#this will return a list of dictionaries values
print(articles)
#and see how we access each one
print(articles[0])
print(articles[0]['title'])
print(articles[0]['link'])
print(articles[0]['votes'])
```

## Now, we need to have the highest number of votes, and print the link and name of the article how do we do that?

Well we have a method call max, that receives an iterable and a key, that key can be a lambda function leets see

```python
response = requests.get('https://news.ycombinator.com/news')
NewSoup = BeautifulSoup(response.text, 'html.parser')
span_tag_text_link = NewSoup.select("span.titleline > a")
#this can be done in just a for loop instaed of a comprehension list
titles = [a.get_text() for a in span_tag_text_link]
links = [a.get(key='href') for a in span_tag_text_link]
#a change is that
#we split the text to just get the first part and convert it to an int
# we have to remember split returns a list, but if we put the exact index we return an element
#with that we will not create a list of lists
#if we do not do that then we are comparing a string for the max value and that will be bad
scores = [int(a.get_text().split()[0]) for a in NewSoup.select(selector='span.score')]
articles = [
    #{}we encapsulate each value in a dictionary
    #first the key, then the value
    {"title": title, "link": link, "votes": vote}
    for title, link, vote in zip(titles, links, scores)
]
```

```python
#LOOK FOR VOTES
print(scores)
```

```python
print(articles)
```

```python
#to get the max voted article 
#we use max(articles, key= lambda arguments: expression
highest_article = max(articles, key=lambda article: article['votes'])
#this is a dictionary
print(highest_article)
```

# Project Movies to watch 

The central idea behind this project is to be able to use BeautifulSoup to obtain some data - like movie titles - from a website like Empire's (or from, say Timeout or Stacker that have curated similar lists). 

```python
import requests
from bs4 import BeautifulSoup
from pathlib import Path

file_path = str(Path.cwd())+'/movies.txt'
URL = "https://web.archive.org/web/20200518073855/https://www.empireonline.com/movies/features/best-movies-2/"
```

```python
# Write your code below this line 👇
response = requests.get(URL)
response.raise_for_status()
data_web = response.text
```

```python
#So seeing the page, the titles are in the tag h3 tha class called title
soup_movies = BeautifulSoup(data_web, 'html.parser')
```

```python
titles = soup_movies.find_all(name='h3', class_='title')
```

```python
#so we know that will return a list with all the data od the tag h3
print(titles)
```

```python
#leets use a list comprehension
titles = [a.get_text() for a in soup_movies.find_all(name='h3', class_='title')]

print(titles)
```

```python
titles = [a.get_text() for a in soup_movies.find_all(name='h3', class_='title')]

print(titles)
#we know that all is in order but we need to inverse it
reversed_list = titles[::-1]
print(reversed_list)
```

```python
#and we can save each item in a file
# we just need to open the file once
with open(file_path, 'a') as file:
    #and for each element we write the content
    for movie in reversed_list:
        file.write(f'{movie}\n')
```
