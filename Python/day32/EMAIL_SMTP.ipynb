## App Project

>---> Project Steps <---

1. Update the birthdays.csv with your friends & family's details. 

2. Check if today matches a birthday in the birthdays.csv

3. If there is a match, pick a random letter (letter_1.txt/letter_2.txt/letter_3.txt) from letter_templates and replace the [NAME] with the person's actual name from birthdays.csv

4. Send the letter generated in step 3 to that person's email address.

```python
from pathlib import Path
import pandas as pd
import datetime as dt
import random
import smtplib
```

```python

#Current working directory
folder_path = Path.cwd()

#letter directory
letter_dir = str(folder_path)+'/data'+'/letter_templates'
#Letters 1-3
letter_one = str(letter_dir)+'/letter_1.txt'
letter_two = str(letter_dir)+'/letter_2.txt'
letter_tree = str(letter_dir)+'/letter_3.txt'

#birthday path
birthday_path = str(folder_path)+'/data'+'/birthdays.csv'
birthday_data = pd.read_csv(birthday_path)

EMAIL = 'tremaine.gulgowski@ethereal.email'
CONTR = 'UuvwygRd9SBuEEf2jN' 
PTT= 587
```

```python
current_time = dt.datetime.now()
current_month = current_time.month
current_day = current_time.day
```

```python
def select_template():
    letter_options = [letter_one,letter_two,letter_tree]
    letter_selected = random.choice(letter_options)
    return letter_selected
```

```python
def replace_name(template,name):
    with open(template, 'r') as file:
        templterr = file.read()
    templterr = templterr.replace('[NAME]',name) 
    return templterr
```

```python
def sent_email(emailToSent, letter):
    with smtplib.SMTP(host= 'smtp.ethereal.email',port= PTT) as connection:
        #we secure the connection
        connection.starttls()
        #we connect ourselfs
        connection.login(user=EMAIL, password=CONTR)

        #For last we sent the email, yeped84053@bitproy.com == row.email
        connection.sendmail(from_addr=EMAIL, to_addrs=emailToSent, 
                            msg='Subject: Congratulations' \
                            f'\n\n{letter}')
```

```python

for index, row in birthday_data.iterrows():
    if row['month'] == current_month and row['day'] == current_day:
        print(f'Birthday, of person: {row['name']}')

        #select a random letter template
        letter_template = select_template()
        #use that for open it and replace the data
        letter_to_sent = replace_name(letter_template, row['name'])
        
        #Maybe we can improve this just save the index and sent the emails to not just wait so long 
        #send the email
        sent_email(row.email, letter_to_sent)
        print('Email sent')
```
